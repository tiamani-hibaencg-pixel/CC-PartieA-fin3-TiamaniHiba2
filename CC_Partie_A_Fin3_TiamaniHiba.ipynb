## 1. Régression linéaire

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Régression linéaire (*Linear Regression*, moindres carrés ordinaires). **Famille :** régression supervisée. **Objectif :** prédire une valeur numérique $y$ à partir de variables explicatives $x_1,\dots,x_p$.
- **Intuition :** on cherche la droite (ou l'hyperplan) qui passe « au plus près » de tous les points d'entraînement.
- **Étapes :** données $(X,y)$ en entrée → calcul des coefficients qui minimisent l'erreur quadratique → prédiction $\hat y$ pour de nouvelles observations.
- **Formule :** $\hat y=\beta_0+\sum_{j=1}^{p}\beta_j x_j$, avec $\hat\beta=\arg\min_\beta\sum_{i=1}^{n}(y_i-\hat y_i)^2=(X^\top X)^{-1}X^\top y$ ; $n$ = nombre d'observations, $p$ = nombre de variables, $\beta_0$ = ordonnée à l'origine, $\beta_j$ = effet de $x_j$.
- **Paramètres appris :** les $\beta_j$. **Hyperparamètres :** pratiquement aucun (seulement le choix d'inclure ou non l'ordonnée à l'origine).
- **Avantages :** (1) très rapide et facile à interpréter ; (2) sert de référence solide pour juger des modèles plus complexes.
- **Limites :** (1) suppose une relation linéaire ; (2) sensible aux valeurs extrêmes et à la forte corrélation entre variables.

### Schéma explicatif

**Comment lire ce schéma.** Les points bleus sont les observations ; la droite rouge est le modèle. Les traits pointillés gris sont les **résidus** (écarts verticaux entre point et droite). La régression choisit la droite qui rend la somme de leurs carrés la plus petite.

```python
sprng=np.random.default_rng(GRAINE);x=np.sort(rng.uniform(0,10,25));y=2+1.5*x+rng.normal(0,2,25)
pente,ord_=np.polyfit(x,y,1);pred=pente*x+ord_
fig,ax=plt.subplots(figsize=(5.2,3.6));ax.scatter(x,y,c='tab:blue',label='observations $(x_i,y_i)$')
ax.plot(x,pred,'r-',label='droite ajustée $\\hat y=\\beta_0+\\beta_1x$');ax.vlines(x,y,pred,colors='gray',linestyles=':',label='résidus $y_i-\\hat y_i$')
ax.set_xlabel('variable explicative x');ax.set_ylabel('cible y');ax.set_title('Régression linéaire : minimiser la somme des résidus au carré');ax.legend(fontsize=8);plt.show()
```

Ce graphique illustre le principe de base de la régression linéaire. Les points bleus sont les observations $(x_i, y_i)$. La ligne rouge représente la droite ajustée $(\hat{y} = \beta_0 + \beta_1 x)$. Les lignes pointillées grises représentent les résidus $(y_i - \hat{y}_i)$, c’est-à-dire l'écart entre la valeur réelle et la valeur prédite par le modèle. L'algorithme cherche à tracer la droite qui minimise la somme des carrés de ces écarts verticaux.   

### Contextes d'utilisation

- **Cas d'usage :** (1) prévoir le chiffre d'affaires d'un magasin à partir du budget publicitaire et de la taille du point de vente (**gestion / marketing**) ; (2) estimer le prix d'un logement selon sa surface.
- **Données et prérequis :** variables numériques (ou encodées), relation à peu près linéaire, peu de colinéarité entre variables ; avoir plus d'observations que de variables.
- **Peu approprié :** prévoir un phénomène qui sature ou explose (par exemple une adoption de produit en courbe en S) : une droite ne peut pas le représenter.
- **Métriques :** MAE, RMSE, $R^2$ ; examen des résidus.

```python
from sklearn.linear_model import LinearRegression
modele=LinearRegression().fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python

- **Entrées :** 5 variables synthétiques (`make_regression`, bruit = 13), 210 observations d'entraînement et 70 de test. **Hyperparamètres :** aucun.
- **Sortie :** MAE sur le test = **9,354**. En moyenne, la prédiction s'écarte de la vraie valeur d'environ 9 unités.
- **Lecture :** le bruit injecté vaut 13 ; l'erreur absolue moyenne minimale attendue est d'environ $13\times0{,}8\approx10{,}4$ ; le modèle s'en approche (l'écart vient de l'échantillon de test).
- **Limite :** les données sont fabriquées avec une relation linéaire : ce résultat favorise mécaniquement ce modèle et ne dit rien de sa performance sur des données réelles.

## 2. Ridge

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Régression Ridge (*Ridge regression*, régularisation L2). **Famille :** régression supervisée. **Objectif :** faire de la régression linéaire en évitant des coefficients démesurés.
- **Intuition :** on ajoute à l'erreur une pénalité proportionnelle à la taille des coefficients ; le modèle préfère alors des coefficients petits et stables.
- **Étapes :** standardiser les variables → minimiser « erreur + pénalité » → prédire.
- **Formule :** $\min_\beta\ \sum_{i=1}^{n}(y_i-\hat y_i)^2+\alpha\sum_{j=1}^{p}\beta_j^2$ ; $\alpha\ge 0$ règle la force de la pénalité ($\alpha=0$ redonne les moindres carrés).
- **Paramètres :** les $\beta_j$. **Hyperparamètre :** $\alpha$ (choisi par validation croisée, ici $\alpha=1$).
- **Avantages :** (1) stable en cas de variables corrélées ; (2) réduit le risque de sur-apprentissage.
- **Limites :** (1) garde toutes les variables (aucune sélection) ; (2) demande de standardiser les variables et d'ajuster $\alpha$.

### Schéma explicatif

**Comment lire ce schéma.** Chaque courbe est un coefficient $\beta_j$ ; l'axe horizontal est la force de pénalisation $\alpha$ (échelle logarithmique). Plus on va vers la droite, plus **tous** les coefficients rétrécissent vers 0, mais ils ne valent jamais exactement 0.

```python
from sklearn.linear_model import Ridge,Lasso
Xs,ys=make_regression(n_samples=200,n_features=8,n_informative=3,noise=10,random_state=GRAINE)
alphas=np.logspace(-2,4,40);coefs=np.array([Ridge(alpha=a).fit(Xs,ys).coef_ for a in alphas])
fig,ax=plt.subplots(figsize=(5.2,3.6));ax.semilogx(alphas,coefs);ax.axhline(0,c='k',lw=.7)
ax.set_xlabel('force de pénalisation $\\alpha$ (échelle log)');ax.set_ylabel('valeur des 8 coefficients $\\beta_j$');ax.set_title('Ridge : les coefficients rétrécissent vers 0\nsans jamais s’annuler exactement');plt.show()
```

Ce graphique montre l'effet de la régression Ridge (régularisation L2). Lorsque la force de pénalisation $\alpha$ augmente (sur l'axe horizontal en échelle logarithmique), les valeurs des 8 coefficients $\beta_j$ rétrécissent progressivement et se rapprochent de 0. Cependant, ils ne s'annulent jamais exactement : les courbes tendent vers 0 sans l'atteindre complètement.

### Contextes d'utilisation

- **Cas d'usage :** (1) prévoir la demande d'un produit à partir de dizaines d'indicateurs économiques très corrélés entre eux (**finance / gestion**) ; (2) modéliser des données génomiques où il y a presque autant de variables que d'observations.
- **Données et prérequis :** variables numériques standardisées ; hyperparamètre $\alpha$ à régler sur un jeu de validation.
- **Peu approprié :** quand on veut éliminer des variables inutiles pour simplifier le modèle (préférer Lasso).
- **Métriques :** MAE, RMSE, $R^2$, courbe de validation en fonction de $\alpha$.

```python
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),Ridge(alpha=1.0)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python

- **Entrées :** mêmes données que l'exemple 1, standardisées dans un `Pipeline`. **Hyperparamètre :** `alpha=1.0`.
- **Sortie :** MAE sur le test = **9,368**, quasiment identique à la régression linéaire (9,35).
- **Lecture :** avec $\alpha=1$, la pénalité est faible et seulement 5 variables sont présentes ; Ridge n'apporte ici pas d'amélioration visible.
- **Limite :** cet exemple ne montre pas l'intérêt de Ridge, qui apparaît surtout avec beaucoup de variables corrélées ; $\alpha$ n'a pas été optimisé.

## 3. Lasso

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Lasso (*Least Absolute Shrinkage and Selection Operator*, régularisation L1). **Famille :** régression supervisée. **Objectif :** prédire une valeur numérique **et** sélectionner automatiquement les variables utiles.
- **Intuition :** la pénalité porte sur la somme des valeurs absolues des coefficients ; elle pousse certains coefficients exactement à zéro.
- **Étapes :** standardiser → minimiser « erreur + pénalité L1 » → ne garder que les variables aux coefficients non nuls → prédire.
- **Formule :** $\min_\beta\ \frac{1}{2n}\sum_{i=1}^{n}(y_i-\hat y_i)^2+\alpha\sum_{j=1}^{p}|\beta_j|$ ; mêmes symboles que Ridge.
- **Paramètres :** les $\beta_j$. **Hyperparamètres :** $\alpha$ (ici 0,1) et `max_iter` (nombre maximal d'itérations de l'optimiseur).
- **Avantages :** (1) produit un modèle parcimonieux, facile à expliquer ; (2) utile quand il y a beaucoup de variables.
- **Limites :** (1) parmi des variables très corrélées, il en choisit une un peu arbitrairement ; (2) biaise les coefficients conservés.

### Schéma explicatif

**Comment lire ce schéma.** Même lecture que pour Ridge, mais ici, quand $\alpha$ augmente, les courbes **touchent exactement 0** l'une après l'autre : les variables correspondantes sont éliminées. Les variables les plus utiles sont celles dont la courbe reste non nulle le plus longtemps.

```python
alphas=np.logspace(-2,2.3,40);coefs=np.array([Lasso(alpha=a,max_iter=20000).fit(Xs,ys).coef_ for a in alphas])
fig,ax=plt.subplots(figsize=(5.2,3.6));ax.semilogx(alphas,coefs);ax.axhline(0,c='k',lw=.7)
ax.set_xlabel('force de pénalisation $\\alpha$ (échelle log)');ax.set_ylabel('valeur des 8 coefficients $\\beta_j$');ax.set_title('Lasso : des coefficients tombent exactement à 0\n(sélection de variables)');plt.show()
```

Ce graphique illustre la régression Lasso (régularisation L1). Contrairement à Ridge, lorsque la force de pénalisation $\alpha$ augmente, certains coefficients tombent exactement à 0. Cela permet d'effectuer une sélection de variables automatique en éliminant les caractéristiques (features) inutiles ou redondantes du modèle.

### Contextes d'utilisation

- **Cas d'usage :** (1) identifier, parmi 200 indicateurs, ceux qui expliquent le mieux le défaut de paiement d'un client (**finance**) ; (2) repérer les gènes les plus liés à une maladie.
- **Données et prérequis :** variables standardisées, nombreuses variables candidates ; $\alpha$ réglé sur validation.
- **Peu approprié :** quand presque toutes les variables sont utiles et corrélées : la sélection devient instable.
- **Métriques :** MAE, RMSE, $R^2$, et nombre de coefficients non nuls.

```python
from sklearn.linear_model import Lasso
modele=make_pipeline(StandardScaler(),Lasso(alpha=.1,max_iter=10000)).fit(Xr_ent,yr_ent)
print('R²:',r2_score(yr_test,modele.predict(Xr_test)))
```

## 4. Régression polynomiale

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Régression polynomiale (*Polynomial regression*). **Famille :** régression supervisée. **Objectif :** modéliser une relation courbe avec un modèle linéaire dans ses coefficients.
- **Intuition :** on ajoute aux variables d'origine leurs carrés et leurs produits ; la régression linéaire travaille ensuite sur ces nouvelles variables.
- **Étapes :** standardiser → créer les termes polynomiaux jusqu'au degré $d$ → ajuster Ridge sur ces termes → prédire.
- **Formule :** pour une variable, $\hat y=\beta_0+\beta_1x+\beta_2x^2+\dots+\beta_dx^d$ ; $d$ = degré du polynôme.
- **Paramètres :** les $\beta_k$. **Hyperparamètres :** le degré $d$ (ici 2) et la pénalité $\alpha$ de Ridge (ici 10).
- **Avantages :** (1) capture des courbures simples ; (2) reste un modèle linéaire facile à ajuster.
- **Limites :** (1) le nombre de termes explose avec le degré et le nombre de variables ; (2) fort risque de sur-apprentissage et mauvaises extrapolations hors du domaine observé.

### Schéma explicatif

**Comment lire ce schéma.** Les points noirs sont les données. La droite orange (degré 1) est trop rigide, la courbe verte (degré 2) suit la tendance, la courbe rouge (degré 12) passe près de chaque point mais oscille : elle a **sur-appris** le bruit.

```python
rng=np.random.default_rng(GRAINE);x=np.sort(rng.uniform(-3,3,40));y=.5*x**2-x+rng.normal(0,.7,40);g=np.linspace(-3,3,200)
fig,ax=plt.subplots(figsize=(5.4,3.6));ax.scatter(x,y,c='k',s=14,label='observations')
for d,c,l in [(1,'tab:orange','degré 1 : sous-apprentissage'),(2,'tab:green','degré 2 : adapté'),(12,'tab:red','degré 12 : sur-apprentissage')]:
    ax.plot(g,np.polyval(np.polyfit(x,y,d),g),c=c,label=l)
ax.set_ylim(-4,8);ax.set_xlabel('x');ax.set_ylabel('y');ax.set_title('Régression polynomiale : choisir le degré');ax.legend(fontsize=8);plt.show()
```

Ce graphique met en évidence le compromis entre biais et variance à travers trois modèles polynomiaux :
Degré 1 (orange) : Modèle linéaire simple, entraînant un sous-apprentissage (underfitting) car il est trop rigide pour capturer la vraie forme des données.
Degré 2 (vert) : Modèle adapté (balanced), qui capture bien la tendance générale sans complexité excessive.
Degré 12 (rouge) : Modèle extrêmement complexe qui passe par tous les points, menant à un sur-apprentissage (overfitting). Il manquera de généralisation sur de nouvelles données.

### Contextes d'utilisation

- **Cas d'usage :** (1) relation entre le prix de vente et le volume vendu, avec rendements décroissants (**marketing**) ; (2) trajectoire d'un véhicule ou courbe d'étalonnage d'un capteur.
- **Données et prérequis :** peu de variables, relation courbe et régulière, degré faible, standardisation.
- **Peu approprié :** prévoir en dehors de la plage observée (un polynôme de haut degré diverge) ou avec beaucoup de variables.
- **Métriques :** MAE, RMSE, $R^2$ ; courbes d'apprentissage pour détecter le sur-apprentissage.

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),PolynomialFeatures(degree=2,include_bias=False),Ridge(alpha=10)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python

- **Entrées :** mêmes 5 variables, étendues au degré 2 (20 termes). **Hyperparamètres :** `degree=2`, `alpha=10`.
- **Sortie :** MAE sur le test = **10,897**, un peu plus élevée que la régression linéaire (9,35).
- **Lecture :** la relation simulée est purement linéaire ; les termes quadratiques n'apportent que du bruit à ajuster, ce qui dégrade légèrement le résultat.
- **Limite :** un seul découpage entraînement/test (70 observations) ne permet pas de dire que l'écart de 1,5 est significatif.

## 5. Arbre de régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Arbre de régression (*Regression tree*, méthode CART). **Famille :** régression supervisée. **Objectif :** prédire une valeur numérique par une suite de questions « variable < seuil ? ».
- **Intuition :** on découpe l'espace en zones rectangulaires où la valeur à prédire est presque constante.
- **Étapes :** choisir la coupure qui réduit le plus l'erreur → recommencer dans chaque zone jusqu'à la profondeur maximale → la prédiction d'une feuille est la moyenne de ses observations.
- **Formule :** critère de coupure $\min_{j,s}\Big[\sum_{x_i\in G}(y_i-\bar y_G)^2+\sum_{x_i\in D}(y_i-\bar y_D)^2\Big]$ ; $j$ = variable, $s$ = seuil, $G$ et $D$ = groupes gauche et droit, $\bar y$ = moyenne du groupe.
- **Paramètres :** variables et seuils des coupures. **Hyperparamètres :** `max_depth` (ici 5), taille minimale des feuilles.
- **Avantages :** (1) interprétable, se lit comme des règles ; (2) gère les relations non linéaires sans standardisation.
- **Limites :** (1) instable (une petite variation des données change l'arbre) ; (2) prédictions en escalier et sur-apprentissage si l'arbre est profond.

### Schéma explicatif

**Comment lire ce schéma.** La courbe rouge est la prédiction : un escalier à 4 marches (4 feuilles). Les droites bleues pointillées sont les **seuils de coupure** choisis par l'arbre ; entre deux seuils, la prédiction est la moyenne des points de la zone.

```python
from sklearn.tree import DecisionTreeRegressor
rng=np.random.default_rng(GRAINE);x=np.sort(rng.uniform(0,6,80));y=np.sin(x)+rng.normal(0,.2,80);g=np.linspace(0,6,400)
arbre=DecisionTreeRegressor(max_depth=2).fit(x[:,None],y)
fig,ax=plt.subplots(figsize=(5.4,3.6));ax.scatter(x,y,s=12,c='gray',label='observations');ax.plot(g,arbre.predict(g[:,None]),'r-',lw=2,label='prédiction (constante par feuille)')
for s in arbre.tree_.threshold[arbre.tree_.feature>=0]: ax.axvline(s,ls='--',c='tab:blue',lw=1)
ax.plot([],[],'--',c='tab:blue',label='seuils de coupure');ax.set_title('Arbre de régression (profondeur 2) : 4 feuilles = 4 paliers');ax.legend(fontsize=8);plt.show()
```

Une forêt aléatoire (Random Forest) combine plusieurs arbres de décision individuels (représentés en lignes bleues pâles, entraînés sur des échantillons bootstrap). La courbe rouge épaisse représente la forêt, c'est-à-dire la moyenne des prédictions de tous les arbres. Cette moyenne permet de lisser les prédictions en réduisant la variance des arbres individuels.

### Contextes d'utilisation

- **Cas d'usage :** (1) estimer le délai de livraison selon la distance, le transporteur et la saison (**logistique**) ; (2) estimer un tarif d'assurance par règles simples.
- **Données et prérequis :** mélange de variables numériques et catégorielles encodées, interactions entre variables ; pas de standardisation nécessaire.
- **Peu approprié :** relation lisse et linéaire (l'arbre l'approche par des marches d'escalier) ou petits échantillons.
- **Métriques :** MAE, RMSE, $R^2$ ; profondeur choisie par validation.

```python
from sklearn.tree import DecisionTreeRegressor
modele=DecisionTreeRegressor(max_depth=5,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python

- **Entrées :** mêmes données. **Hyperparamètres :** `max_depth=5`, `random_state=42`.
- **Sortie :** MAE sur le test = **37,717**, soit environ quatre fois celle de la régression linéaire (9,35).
- **Lecture :** avec 210 observations d'entraînement et une relation linéaire entre 5 variables, un seul arbre l'approche mal par paliers.
- **Limite :** ce constat vaut pour ces données simulées ; sur des données à interactions non linéaires, un arbre peut être compétitif.

## 6. Forêt aléatoire — régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Forêt aléatoire pour la régression (*Random Forest Regressor*). **Famille :** régression supervisée (méthode d'ensemble, *bagging*). **Objectif :** réduire l'instabilité d'un arbre en en moyennant beaucoup.
- **Intuition :** on entraîne de nombreux arbres sur des échantillons tirés avec remise et sur des sous-ensembles de variables ; leurs erreurs différentes se compensent.
- **Étapes :** tirer un échantillon *bootstrap* → construire un arbre avec, à chaque coupure, un sous-ensemble aléatoire de variables → répéter $B$ fois → moyenner.
- **Formule :** $\hat y(x)=\frac{1}{B}\sum_{b=1}^{B}T_b(x)$ ; $B$ = nombre d'arbres, $T_b$ = le $b$-ième arbre.
- **Paramètres :** les arbres. **Hyperparamètres :** `n_estimators` ($B$=80), `max_depth`, `max_features`, taille minimale des feuilles.
- **Avantages :** (1) robuste et performant sans beaucoup de réglages ; (2) donne l'importance des variables.
- **Limites :** (1) peu interprétable et plus lent qu'un arbre ; (2) ne sait pas extrapoler hors de la plage des données d'entraînement.

### Schéma explicatif

**Comment lire ce schéma.** Chaque courbe bleue claire est un arbre entraîné sur un échantillon *bootstrap* : elles sont irrégulières et différentes. La courbe rouge épaisse est leur **moyenne** (la forêt) : plus lisse et plus stable que chaque arbre.

```python
rng=np.random.default_rng(GRAINE);x=np.sort(rng.uniform(0,6,100));y=np.sin(x)+rng.normal(0,.3,100);g=np.linspace(0,6,300)
preds=[]
fig,ax=plt.subplots(figsize=(5.6,3.7));ax.scatter(x,y,s=10,c='gray',label='observations')
for b in range(8):
    idx=rng.integers(0,100,100);p=DecisionTreeRegressor(max_depth=4).fit(x[idx,None],y[idx]).predict(g[:,None]);preds.append(p);ax.plot(g,p,lw=.7,alpha=.5,c='tab:blue')
ax.plot([],[],c='tab:blue',alpha=.5,label='arbres individuels (échantillons bootstrap)');ax.plot(g,np.mean(preds,axis=0),'r-',lw=2.5,label='forêt = moyenne des arbres')
ax.set_title('Forêt aléatoire (régression) : la moyenne lisse les arbres');ax.legend(fontsize=8);plt.show()
```

### Contextes d'utilisation

- **Cas d'usage :** (1) prévoir les ventes d'un produit avec des dizaines de variables mêlées (promotions, météo, prix) (**commerce**) ; (2) estimer la valeur d'un bien immobilier.
- **Données et prérequis :** tableaux de taille moyenne, variables mixtes, relations non linéaires ; pas de standardisation.
- **Peu approprié :** extrapoler une tendance (la prévision reste plate au-delà des valeurs vues) ou nécessité d'un modèle explicable ligne par ligne.
- **Métriques :** MAE, RMSE, $R^2$, erreur *out-of-bag*.

```python
from sklearn.ensemble import RandomForestRegressor
modele=RandomForestRegressor(n_estimators=80,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python

- **Entrées :** mêmes données. **Hyperparamètres :** `n_estimators=80`, `random_state=42`.
- **Sortie :** MAE sur le test = **24,374**, nettement meilleure que l'arbre seul (37,7) mais moins bonne que la régression linéaire (9,35).
- **Lecture :** la moyenne de 80 arbres stabilise la prédiction, ce qui confirme l'intérêt du *bagging* par rapport à un arbre unique.
- **Limite :** sur des données dont la relation est linéaire, les modèles linéaires gardent l'avantage ; cet exemple ne mesure pas la performance sur données réelles.

## 7. Gradient Boosting — régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Gradient Boosting pour la régression (*Gradient Boosting Regressor*, version `HistGradientBoostingRegressor` de scikit-learn). **Famille :** régression supervisée (méthode d'ensemble, *boosting*). **Objectif :** obtenir un modèle très précis en empilant de petits arbres.
- **Intuition :** chaque nouvel arbre apprend à corriger les erreurs (résidus) laissées par les arbres précédents.
- **Étapes :** prédiction initiale (moyenne) → calcul des résidus → ajustement d'un petit arbre sur ces résidus → ajout pondéré → répétition. La version « Hist » regroupe les valeurs des variables en classes pour accélérer le calcul.
- **Formule :** $F_m(x)=F_{m-1}(x)+\nu\,h_m(x)$ ; $F_m$ = modèle après $m$ arbres, $h_m$ = arbre ajusté sur les résidus, $\nu$ = taux d'apprentissage.
- **Paramètres :** les arbres $h_m$. **Hyperparamètres :** `max_iter` (nombre d'arbres, ici 80), `learning_rate` ($\nu$), profondeur, `random_state`.
- **Avantages :** (1) souvent parmi les meilleurs modèles sur données tabulaires ; (2) gère les relations non linéaires et les valeurs manquantes.
- **Limites :** (1) sensible au sur-apprentissage si mal réglé ; (2) entraînement séquentiel et moins interprétable.

### Schéma explicatif

**Comment lire ce schéma.** Trois étapes du boosting (1, 5 et 30 petits arbres). La courbe rouge est le modèle courant : à gauche il est très grossier, puis chaque arbre ajouté corrige les résidus et la courbe se rapproche des points.

```python
rng=np.random.default_rng(GRAINE);x=np.sort(rng.uniform(0,6,100));y=np.sin(x)+rng.normal(0,.2,100);g=np.linspace(0,6,300)
F=np.full(100,y.mean());Fg=np.full(300,y.mean());nu=.5;fig,axs=plt.subplots(1,3,figsize=(11,3.2),sharey=True);etapes={1:0,5:1,30:2}
for m in range(1,31):
    h=DecisionTreeRegressor(max_depth=2).fit(x[:,None],y-F);F=F+nu*h.predict(x[:,None]);Fg=Fg+nu*h.predict(g[:,None])
    if m in etapes:
        a=axs[etapes[m]];a.scatter(x,y,s=8,c='gray');a.plot(g,Fg,'r-',lw=2);a.set_title(f'après {m} arbre(s)')
fig.suptitle('Gradient Boosting : chaque petit arbre corrige les résidus du modèle précédent',y=1.02);plt.show()
```

Le Gradient Boosting est une méthode d'ensemble séquentielle. Chaque petit arbre ajouté corrige les erreurs (résidus) commises par le modèle précédent. On observe l'évolution du modèle : après 1 arbre (trop simple), après 5 arbres (ajustement intermédiaire), et après 30 arbres (modèle précis qui épouse finement la forme sinusoïdale des données).

### Contextes d'utilisation

- **Cas d'usage :** (1) prévoir le montant de sinistres ou la perte attendue d'un portefeuille de crédits (**finance**) ; (2) prévoir la demande de livraison par zone (**logistique**).
- **Données et prérequis :** données tabulaires de taille moyenne à grande ; réglage du taux d'apprentissage et du nombre d'arbres sur validation.
- **Peu approprié :** très petits échantillons (sur-apprentissage) ou besoin de décisions entièrement transparentes.
- **Métriques :** MAE, RMSE, $R^2$ ; courbe d'erreur de validation selon le nombre d'itérations.

```python
from sklearn.ensemble import HistGradientBoostingRegressor
modele=HistGradientBoostingRegressor(max_iter=80,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python

- **Entrées :** mêmes données. **Hyperparamètres :** `max_iter=80`, autres par défaut.
- **Sortie :** MAE sur le test = **23,266**, du même ordre que la forêt (24,4) et bien au-dessus de la régression linéaire (9,35).
- **Lecture :** le boosting apprend par paliers ; avec 210 observations et une relation linéaire, il manque de données pour la reconstituer finement.
- **Limite :** aucun réglage n'a été fait (pas de validation) ; la comparaison avec le modèle linéaire reflète la nature simulée des données.

## 8. Régression logistique

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Régression logistique (*Logistic regression*). **Famille :** classification supervisée. **Objectif :** estimer la probabilité qu'une observation appartienne à la classe 1.
- **Intuition :** on calcule un score linéaire, puis la fonction sigmoïde le transforme en probabilité entre 0 et 1.
- **Étapes :** standardiser → ajuster les coefficients en minimisant la perte logarithmique → calculer $P(y=1\mid x)$ → classer en 1 si la probabilité dépasse un seuil (0,5 par défaut).
- **Formule :** $P(y=1\mid x)=\sigma(z)=\dfrac{1}{1+e^{-z}}$ avec $z=\beta_0+\sum_j\beta_jx_j$ ; perte $-\sum_i\big[y_i\ln p_i+(1-y_i)\ln(1-p_i)\big]$, $p_i$ = probabilité prédite pour l'observation $i$.
- **Paramètres :** les $\beta_j$. **Hyperparamètres :** `C` (inverse de la force de régularisation), seuil de décision, `max_iter`.
- **Avantages :** (1) fournit des probabilités interprétables (les coefficients donnent des *odds ratios*) ; (2) rapide et solide comme modèle de référence.
- **Limites :** (1) frontière de décision linéaire ; (2) sensible aux variables corrélées et aux valeurs extrêmes.

### Schéma explicatif

**Comment lire ce schéma.** Les points sont les observations (classe 0 en bleu, classe 1 en rouge) ; la courbe noire est la probabilité estimée $P(y=1\mid x)$. Là où elle franchit le seuil pointillé de 0,5, la prédiction passe de la classe 0 à la classe 1.

```python
from sklearn.linear_model import LogisticRegression
rng=np.random.default_rng(GRAINE);x=np.r_[rng.normal(-1,1,40),rng.normal(1.5,1,40)];y=np.r_[np.zeros(40),np.ones(40)];g=np.linspace(-4,5,200)
m=LogisticRegression().fit(x[:,None],y);fig,ax=plt.subplots(figsize=(5.4,3.5))
ax.scatter(x,y,c=y,cmap='coolwarm',edgecolor='k',s=22,label='observations (classe 0 ou 1)');ax.plot(g,m.predict_proba(g[:,None])[:,1],'k-',lw=2,label='$P(y=1\\mid x)=\\sigma(\\beta_0+\\beta_1x)$')
ax.axhline(.5,ls='--',c='gray');ax.set_xlabel('x');ax.set_ylabel('probabilité');ax.text(-3.9,.53,'seuil 0,5',fontsize=8);ax.set_title('Régression logistique : la sigmoïde transforme un score en probabilité');ax.legend(fontsize=8,loc='center right');plt.show()
```

Utilisée pour la classification binaire (classe 0 ou 1), la régression logistique applique une fonction sigmoïde pour transformer un score linéaire en une probabilité comprise entre 0 et 1. Avec un seuil de 0,5 (ligne horizontale pointillée), tout point situé au-dessus est classé dans une catégorie (ex: rouge/1) et tout point en dessous dans l'autre (ex: bleu/0).

### Contextes d'utilisation

- **Cas d'usage :** (1) estimer la probabilité de défaut de paiement d'un client (**finance**, score de crédit) ; (2) prédire si un client va se désabonner (**marketing**).
- **Données et prérequis :** variables numériques ou encodées, standardisation, classes pas trop déséquilibrées (sinon ajuster le seuil ou les poids).
- **Peu approprié :** frontière très non linéaire entre les classes (par exemple deux cercles concentriques).
- **Métriques :** F1, précision, rappel, ROC-AUC, PR-AUC (si classes déséquilibrées), log-loss.

```python
from sklearn.linear_model import LogisticRegression
modele=make_pipeline(StandardScaler(),LogisticRegression(max_iter=1000,random_state=GRAINE)).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python

- **Entrées :** 6 variables synthétiques, 300 observations d'entraînement et 100 de test, classes équilibrées par stratification. **Hyperparamètres :** `max_iter=1000` (le reste par défaut).
- **Sortie :** F1 sur le test = **0,811**.
- **Lecture :** un F1 d'environ 0,81 est le deuxième plus faible de la série, juste devant Naive Bayes (0,755) ; la frontière linéaire de ce modèle est sans doute trop simple pour ces données.
- **Limite :** avec 100 observations de test, une différence de quelques points de F1 (3 à 4 observations) n'est pas statistiquement concluante ; aucune validation croisée n'est faite.

## 9. KNN

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** K plus proches voisins (*k-Nearest Neighbors*, KNN). **Famille :** classification supervisée (aussi utilisable en régression). **Objectif :** classer une observation selon la classe majoritaire de ses voisins.
- **Intuition :** « dis-moi qui sont tes voisins, je te dirai qui tu es ».
- **Étapes :** standardiser → pour un nouveau point, calculer sa distance à tous les points d'entraînement → retenir les $k$ plus proches → prendre le vote majoritaire.
- **Formule :** distance euclidienne $d(x,x')=\sqrt{\sum_{j=1}^{p}(x_j-x'_j)^2}$ ; la classe prédite est celle qui apparaît le plus parmi les $k$ voisins.
- **Paramètres :** aucun (le modèle mémorise les données). **Hyperparamètres :** $k$ (ici 5), distance, pondération des voisins.
- **Avantages :** (1) simple, sans phase d'apprentissage ; (2) s'adapte à des frontières irrégulières.
- **Limites :** (1) prédiction lente sur de gros jeux de données ; (2) très sensible à l'échelle des variables et aux variables inutiles.

### Schéma explicatif

**Comment lire ce schéma.** L'étoile verte est le point à classer, les cercles dorés ses 5 plus proches voisins, le cercle pointillé le rayon qui les contient. La classe prédite est celle qui domine parmi ces 5 voisins.

```python
from sklearn.neighbors import NearestNeighbors
Xb,yb=make_blobs(n_samples=60,centers=[[0,0],[2.2,1.5]],cluster_std=1.0,random_state=GRAINE);q=np.array([[1.1,.7]])
idx=NearestNeighbors(n_neighbors=5).fit(Xb).kneighbors(q)[1][0];rayon=np.linalg.norm(Xb[idx[-1]]-q[0])
fig,ax=plt.subplots(figsize=(5,4.2));ax.scatter(Xb[:,0],Xb[:,1],c=yb,cmap='coolwarm',edgecolor='k',s=25)
ax.scatter(Xb[idx,0],Xb[idx,1],s=150,facecolors='none',edgecolors='gold',lw=2,label='5 plus proches voisins')
ax.scatter(*q[0],marker='*',s=250,c='lime',edgecolor='k',label='point à classer');ax.add_patch(plt.Circle(q[0],rayon,fill=False,ls='--'))
ax.set_aspect('equal');ax.set_title('KNN (k=5) : vote majoritaire des voisins');ax.legend(fontsize=8);plt.show()
```

L'algorithme KNN (avec $K=5$ sur ce schéma) classe un nouveau point (l'étoile verte) en regardant ses 5 plus proches voisins dans l'espace (entourés en cercles jaunes). Il attribue au point la classe majoritaire parmi ces voisins par un système de vote (ici, la classe rouge l'emporte).   

### Contextes d'utilisation

- **Cas d'usage :** (1) recommander des produits en s'appuyant sur des clients au profil proche (**marketing**) ; (2) classer des activités (marche, course) d'après des capteurs.
- **Données et prérequis :** peu de variables, toutes standardisées, taille de données modérée.
- **Peu approprié :** données de grande dimension (malédiction de la dimension) ou très volumineuses.
- **Métriques :** F1, rappel, précision, ROC-AUC ; $k$ choisi par validation.

```python
from sklearn.neighbors import KNeighborsClassifier
modele=make_pipeline(StandardScaler(),KNeighborsClassifier(n_neighbors=5)).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

## 10. Naive Bayes

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Naive Bayes gaussien (*Gaussian Naive Bayes*). **Famille :** classification supervisée probabiliste. **Objectif :** classer en appliquant le théorème de Bayes.
- **Intuition :** pour chaque classe, on apprend la loi (gaussienne) de chaque variable ; on choisit la classe la plus vraisemblable pour l'observation.
- **Étapes :** estimer la proportion de chaque classe et la moyenne et l'écart-type de chaque variable par classe → pour un nouvel $x$, calculer un score par classe → prendre le plus grand.
- **Formule :** $P(c\mid x)\propto P(c)\prod_{j=1}^{p}P(x_j\mid c)$ avec $P(x_j\mid c)=\mathcal N(x_j;\mu_{jc},\sigma_{jc}^2)$ ; $c$ = classe, $\mu_{jc}$ et $\sigma_{jc}$ = moyenne et écart-type de la variable $j$ dans la classe $c$.
- **Paramètres :** proportions, $\mu_{jc}$ et $\sigma_{jc}$. **Hyperparamètres :** presque aucun (lissage de la variance `var_smoothing`).
- **Avantages :** (1) extrêmement rapide, même avec peu de données ; (2) fonctionne bien en grande dimension (texte).
- **Limites :** (1) suppose les variables indépendantes entre elles dans chaque classe, rarement vrai ; (2) probabilités souvent mal calibrées.

### Schéma explicatif

**Comment lire ce schéma.** Chaque cloche est la loi d'une variable dans une classe (bleue : A, rouge : B). Pour le nouvel exemple (droite pointillée), on lit la hauteur de chaque cloche (points) : la plus haute, pondérée par la proportion de la classe, l'emporte.

```python
g=np.linspace(-4,7,300);dens=lambda x,m,s:np.exp(-(x-m)**2/(2*s**2))/(s*np.sqrt(2*np.pi));x0=1.5
fig,ax=plt.subplots(figsize=(5.6,3.6));ax.plot(g,dens(g,0,1),c='tab:blue',label='$P(x\\mid$classe A$)$ : gaussienne apprise');ax.plot(g,dens(g,3,1.3),c='tab:red',label='$P(x\\mid$classe B$)$ : gaussienne apprise')
ax.axvline(x0,ls='--',c='k');ax.plot([x0,x0],[0,dens(x0,0,1)],'o',c='tab:blue');ax.plot([x0,x0],[0,dens(x0,3,1.3)],'o',c='tab:red')
ax.text(x0+.1,.36,'nouvel exemple $x$',fontsize=8);ax.set_xlabel('valeur d’une variable');ax.set_ylabel('densité');ax.set_title('Naive Bayes : on compare $P(\\mathrm{classe})\\times P(x\\mid\\mathrm{classe})$');ax.legend(fontsize=8);plt.show()
```

Le classifieur Naïve Bayes modélise la distribution des données pour chaque classe sous forme de lois normales (gaussiennes). Pour un nouvel exemple $x$ (ligne verticale noire pointillée), le modèle compare la densité de probabilité estimée pour la classe A (courbe bleue) et pour la classe B (courbe rouge) afin de déterminer quelle classe a la plus forte probabilité d'avoir généré cette observation.   

### Contextes d'utilisation

- **Cas d'usage :** (1) filtrer des courriels indésirables ou classer des avis clients par thème (**marketing**) ; (2) diagnostic rapide à partir de quelques mesures.
- **Données et prérequis :** variables à peu près indépendantes et de forme proche d'une loi normale (version gaussienne).
- **Peu approprié :** variables fortement corrélées (par exemple deux mesures redondantes du même phénomène) : la même information est comptée deux fois.
- **Métriques :** F1, précision, rappel, ROC-AUC ; qualité de calibration des probabilités.

```python
from sklearn.naive_bayes import GaussianNB
modele=GaussianNB().fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python

- **Entrées :** mêmes données, sans standardisation (inutile ici). **Hyperparamètres :** valeurs par défaut.
- **Sortie :** F1 sur le test = **0,755**, le plus faible des modèles testés.
- **Lecture :** une explication possible est que l'hypothèse d'indépendance ou de normalité des variables n'est pas respectée ; il faudrait l'examiner (corrélations, histogrammes) avant de conclure.
- **Limite :** l'écart avec les autres modèles repose sur 100 observations de test ; la cause n'a pas été vérifiée.

## 11. Arbre de décision

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Arbre de décision (*Decision tree classifier*, CART). **Famille :** classification supervisée. **Objectif :** classer par une suite de questions simples sur les variables.
- **Intuition :** on pose à chaque étape la question qui sépare le mieux les classes, comme dans un jeu des vingt questions.
- **Étapes :** choisir la variable et le seuil qui rendent les deux groupes les plus « purs » → répéter dans chaque groupe jusqu'à `max_depth` → la classe d'une feuille est la classe majoritaire de ses observations.
- **Formule :** impureté de Gini $G=1-\sum_{k}p_k^2$, où $p_k$ = proportion de la classe $k$ dans un nœud ; la coupure retenue maximise $G_{parent}-\sum_{enfant}\frac{n_{enfant}}{n}G_{enfant}$.
- **Paramètres :** variables et seuils. **Hyperparamètres :** `max_depth` (ici 4), `min_samples_leaf`, critère (Gini ou entropie).
- **Avantages :** (1) règles lisibles par un non-spécialiste ; (2) aucune standardisation requise.
- **Limites :** (1) instable et prompt au sur-apprentissage ; (2) frontières en rectangles, peu adaptées aux frontières obliques.

### Schéma explicatif

**Comment lire ce schéma.** À gauche, une seule coupure sépare le plan en 2 zones ; à droite, 4 niveaux de coupures créent des zones rectangulaires. Les couleurs de fond sont la classe prédite ; les points sont les données d'entraînement.

```python
from sklearn.tree import DecisionTreeClassifier
def frontiere(ax,modele,X,y,titre=''):
    xx,yy=np.meshgrid(np.linspace(X[:,0].min()-.6,X[:,0].max()+.6,200),np.linspace(X[:,1].min()-.6,X[:,1].max()+.6,200))
    zz=modele.predict(np.c_[xx.ravel(),yy.ravel()]).reshape(xx.shape);ax.contourf(xx,yy,zz,alpha=.25,cmap='coolwarm')
    ax.scatter(X[:,0],X[:,1],c=y,cmap='coolwarm',edgecolor='k',s=22);ax.set_title(titre,fontsize=9)
fig,axs=plt.subplots(1,2,figsize=(9,3.8))
frontiere(axs[0],DecisionTreeClassifier(max_depth=1).fit(X_lune,y_lune),X_lune,y_lune,'profondeur 1 : une seule coupure')
frontiere(axs[1],DecisionTreeClassifier(max_depth=4).fit(X_lune,y_lune),X_lune,y_lune,'profondeur 4 : zones rectangulaires')
fig.suptitle('Arbre de décision : l’espace est découpé par des coupures parallèles aux axes');plt.show()
```

### Contextes d'utilisation

- **Cas d'usage :** (1) décider d'accorder ou non un crédit selon des règles lisibles (**finance**) ; (2) trier des demandes de support client selon quelques critères.
- **Données et prérequis :** variables mixtes, besoin d'explicabilité ; profondeur limitée.
- **Peu approprié :** quand on vise la meilleure précision possible : un seul arbre est généralement dépassé par les ensembles.
- **Métriques :** F1, précision, rappel, ROC-AUC ; taille de l'arbre.

```python
from sklearn.tree import DecisionTreeClassifier
modele=DecisionTreeClassifier(max_depth=4,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python

- **Entrées :** mêmes données de classification. **Hyperparamètres :** `max_depth=4`.
- **Sortie :** F1 sur le test = **0,891**.
- **Lecture :** un arbre simple de profondeur 4 fait mieux que la régression logistique (0,811) et reste interprétable.
- **Limite :** sa performance dépend fortement du découpage entraînement/test ; elle est plus instable que celle des ensembles.

## 12. Forêt aléatoire — classification

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Forêt aléatoire pour la classification (*Random Forest Classifier*). **Famille :** classification supervisée (ensemble, *bagging*). **Objectif :** stabiliser et améliorer la prédiction d'un arbre de décision.
- **Intuition :** de nombreux arbres, différents car entraînés sur des échantillons et des variables différents, votent ; la majorité est plus fiable qu'un seul arbre.
- **Étapes :** pour $b=1\dots B$ : échantillon *bootstrap* + arbre avec sous-ensemble aléatoire de variables à chaque coupure → vote des $B$ arbres.
- **Formule :** $\hat c(x)=\operatorname{mode}\{T_1(x),\dots,T_B(x)\}$ ; $T_b$ = le $b$-ième arbre, $B$ = nombre d'arbres.
- **Paramètres :** les arbres. **Hyperparamètres :** `n_estimators` ($B$=80), `max_depth`, `max_features`, `class_weight`.
- **Avantages :** (1) très bonne performance sans réglage fin ; (2) mesure l'importance des variables.
- **Limites :** (1) modèle opaque et plus coûteux en mémoire ; (2) probabilités parfois mal calibrées.

### Schéma explicatif

**Comment lire ce schéma.** Les trois premiers graphiques sont des arbres individuels, entraînés sur des échantillons différents : leurs frontières sont irrégulières et différentes. Le dernier est le **vote** de 100 arbres : sa frontière est plus régulière.

```python
from sklearn.ensemble import RandomForestClassifier
rng=np.random.default_rng(GRAINE);fig,axs=plt.subplots(1,4,figsize=(13,3.3))
for k in range(3):
    idx=rng.integers(0,len(X_lune),len(X_lune));frontiere(axs[k],DecisionTreeClassifier(max_depth=4,max_features=1,random_state=k).fit(X_lune[idx],y_lune[idx]),X_lune,y_lune,f'arbre {k+1} (bootstrap)')
frontiere(axs[3],RandomForestClassifier(n_estimators=100,max_depth=4,random_state=GRAINE).fit(X_lune,y_lune),X_lune,y_lune,'forêt : vote de 100 arbres')
fig.suptitle('Forêt aléatoire : des arbres instables et différents, un vote stable');plt.show()
```

### Contextes d'utilisation

- **Cas d'usage :** (1) détecter les clients susceptibles de faire défaut ou de quitter l'entreprise (**finance / marketing**) ; (2) classer des images ou des signaux à partir de descripteurs.
- **Données et prérequis :** données tabulaires, variables mixtes, possibilité de classes déséquilibrées (utiliser `class_weight`).
- **Peu approprié :** données à très haute dimension creuses (texte brut) ou exigence d'explication totale.
- **Métriques :** F1, rappel, précision, ROC-AUC, PR-AUC ; erreur *out-of-bag*.

```python
from sklearn.ensemble import RandomForestClassifier
modele=RandomForestClassifier(n_estimators=80,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python

- **Entrées :** mêmes données. **Hyperparamètres :** `n_estimators=80`, `random_state=42`.
- **Sortie :** F1 sur le test = **0,918**, meilleur que l'arbre seul (0,891).
- **Lecture :** le vote de 80 arbres améliore l'arbre isolé de quelques points.
- **Limite :** l'écart (environ 3 points de F1) correspond à très peu d'observations de test ; il n'est pas démontré par ce seul découpage.

## 13. SVM

- **Formule :** $\min_{w,b}\ \tfrac12\|w\|^2+C\sum_i\xi_i$ sous $y_i(w\cdot x_i+b)\ge 1-\xi_i$ ; $w$ = vecteur normal, $b$ = décalage, $\xi_i$ = marge violée, $C$ = tolérance aux erreurs ; noyau RBF $K(x,x')=e^{-\gamma\|x-x'\|^2}$.

- **Avantages :** (1) efficace en grande dimension ; (2) le noyau permet des frontières non linéaires.
- **Limites :** (1) coûteux sur de grands jeux de données (plus de 10 000 observations) ; (2) pas de probabilités directes et choix de $C$ et $\gamma$ délicat.

**Comment lire ce schéma.** La droite pleine est l'hyperplan séparateur, les pointillés sont les marges. Les points entourés d'or (**vecteurs de support**) sont les seuls à déterminer la frontière. Plus la marge est large, mieux c'est.

## 15. AdaBoost

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** AdaBoost (*Adaptive Boosting*). **Famille :** classification supervisée (ensemble, *boosting*). **Objectif :** combiner beaucoup de classifieurs très simples (des « souches » d'arbres à une coupure) en un classifieur fort.
- **Intuition :** à chaque tour, on augmente le poids des observations mal classées pour forcer le classifieur suivant à s'y concentrer.
- **Étapes :** poids égaux → entraîner une souche pondérée → calculer son erreur $\varepsilon_t$ → augmenter le poids des erreurs → répéter → vote pondéré.
- **Formule :** poids du classifieur $\alpha_t=\tfrac12\ln\frac{1-\varepsilon_t}{\varepsilon_t}$ ; prédiction finale $H(x)=\operatorname{signe}\big(\sum_{t=1}^{T}\alpha_t h_t(x)\big)$ ; $h_t$ = souche du tour $t$, $T$ = nombre de tours.
- **Paramètres :** les souches et leurs poids $\alpha_t$. **Hyperparamètres :** `n_estimators` (ici 70), `learning_rate`, classifieur de base.
- **Avantages :** (1) simple, peu de réglages ; (2) bon sur des frontières complexes avec des classifieurs faibles.
- **Limites :** (1) sensible au bruit et aux étiquettes erronées (elles gagnent du poids) ; (2) souvent dépassé par le Gradient Boosting.

### Schéma explicatif

**Comment lire ce schéma.** Chaque graphique est un tour : la frontière est celle d'une souche (une seule coupure) et la taille des cercles noirs est le **poids** des points. Les points mal classés au tour précédent grossissent et pèsent plus au tour suivant ; $\alpha$ mesure la confiance dans la souche.

- **Formule :** objectif $\sum_i L(y_i,\hat y_i)+\sum_k\Omega(f_k)$ avec $\Omega(f)=\gamma T+\tfrac12\lambda\sum_j w_j^2$ ; $T$ = nombre de feuilles, $w_j$ = poids de la feuille $j$, $\gamma$ et $\lambda$ = pénalités.

- **Avantages :** (1) très bonne performance et rapide (parallélisme) ; (2) régularisation intégrée et gestion des valeurs manquantes.

## 17. K-means

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** K-means (*K-means clustering*, k-moyennes). **Famille :** clustering (apprentissage non supervisé). **Objectif :** regrouper des observations en $k$ groupes compacts, sans étiquettes.
- **Intuition :** chaque groupe est représenté par son centre (centroïde) ; chaque point rejoint le centre le plus proche.
- **Étapes :** choisir $k$ centres initiaux (`k-means++`) → affecter chaque point au centre le plus proche → recalculer chaque centre comme la moyenne de son groupe → répéter jusqu'à stabilité.
- **Formule :** minimiser l'inertie $J=\sum_{k=1}^{K}\sum_{x_i\in C_k}\|x_i-\mu_k\|^2$ ; $C_k$ = groupe $k$, $\mu_k$ = son centroïde.
- **Paramètres :** les centroïdes $\mu_k$ et les affectations. **Hyperparamètres :** $K$ (ici 3), `n_init` (nombre de relances, ici 10), méthode d'initialisation.
- **Avantages :** (1) simple, rapide et passe à l'échelle ; (2) résultat facile à interpréter par les centres.
- **Limites :** (1) il faut fixer $K$ à l'avance et les groupes sont supposés ronds et de tailles similaires ; (2) sensible à l'initialisation et aux valeurs extrêmes.

### Schéma explicatif

**Comment lire ce schéma.** Chaque graphique est une étape. Les croix rouges sont les centroïdes, les couleurs les groupes. On voit les centroïdes bouger vers le centre de leur groupe après chaque mise à jour, jusqu'à se stabiliser.

```python
rng=np.random.default_rng(1);C=X_clu[rng.choice(len(X_clu),3,replace=False)];fig,axs=plt.subplots(1,3,figsize=(11,3.4))
for it,a in enumerate(axs):
    lab=((X_clu[:,None]-C[None])**2).sum(2).argmin(1);a.scatter(X_clu[:,0],X_clu[:,1],c=lab,cmap='viridis',s=10,alpha=.7);a.scatter(C[:,0],C[:,1],marker='X',s=170,c='red',edgecolor='k')
    a.set_title(['initialisation (centroïdes au hasard)','après 1 mise à jour','après 2 mises à jour'][it],fontsize=9);C=np.array([X_clu[lab==k].mean(0) if (lab==k).any() else C[k] for k in range(3)])
fig.suptitle('K-means : (1) affecter chaque point au centroïde le plus proche, (2) recalculer les centroïdes (X rouges)');plt.show()
```

### Contextes d'utilisation

- **Cas d'usage :** (1) segmenter la clientèle selon récence, fréquence et montant d'achat (**marketing**) ; (2) regrouper des points de livraison pour organiser des tournées (**logistique**).
- **Données et prérequis :** variables numériques standardisées, groupes plutôt compacts.
- **Peu approprié :** groupes de forme allongée ou imbriquée (par exemple deux croissants) ou données avec beaucoup de bruit.
- **Métriques :** silhouette, inertie (méthode du coude), stabilité entre initialisations ; indice de Rand ajusté si des étiquettes réelles existent.

```python
from sklearn.cluster import KMeans
modele=KMeans(n_clusters=3,n_init=10,random_state=GRAINE).fit(X_clu)
print('Silhouette:',silhouette_score(X_clu,modele.labels_))
plt.scatter(X_clu[:,0],X_clu[:,1],c=modele.labels_,s=12);plt.title('K-means');plt.show()
```

## 18. Classification hiérarchique

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Classification hiérarchique ascendante (*Agglomerative hierarchical clustering*). **Famille :** clustering. **Objectif :** construire une hiérarchie de regroupements, représentée par un dendrogramme.
- **Intuition :** au départ chaque point est son propre groupe ; on fusionne à chaque étape les deux groupes les plus proches.
- **Étapes :** calculer les distances → fusionner les deux groupes les plus proches selon le critère de liaison → répéter jusqu'à un seul groupe → couper l'arbre à la hauteur voulue.
- **Formule :** critère de Ward (par défaut dans scikit-learn) : on fusionne les groupes $A$ et $B$ qui minimisent l'augmentation de variance interne $\Delta=\frac{|A||B|}{|A|+|B|}\|\mu_A-\mu_B\|^2$ ; $|A|$ = effectif, $\mu_A$ = centre.
- **Paramètres :** aucun appris. **Hyperparamètres :** `n_clusters` (ici 3), `linkage` (ward, complete, average, single), distance.
- **Avantages :** (1) pas besoin de fixer $K$ avant de voir le dendrogramme ; (2) révèle une structure à plusieurs niveaux.
- **Limites :** (1) coût au moins quadratique en mémoire et en temps (limité à quelques milliers de points) ; (2) une fusion faite n'est jamais défaite.

### Schéma explicatif

**Comment lire ce schéma.** Chaque feuille du bas est un point ; plus deux branches se rejoignent haut, plus les groupes sont éloignés. La ligne pointillée horizontale coupe l'arbre : le nombre de branches coupées donne le nombre de groupes (3 couleurs).

```python
from scipy.cluster.hierarchy import linkage,dendrogram
sous=X_clu[np.random.default_rng(GRAINE).choice(len(X_clu),20,replace=False)];Z=linkage(sous,'ward')
fig,ax=plt.subplots(figsize=(6.4,3.6));dendrogram(Z,ax=ax,color_threshold=Z[-2,2]+.01);ax.axhline(Z[-2,2]+.01,ls='--',c='k')
ax.set_xlabel('observations (20 points tirés au hasard)');ax.set_ylabel('distance de fusion (Ward)');ax.set_title('Dendrogramme : couper à la ligne pointillée donne 3 groupes');plt.show()
```

### Contextes d'utilisation

- **Cas d'usage :** (1) construire une taxonomie de produits ou de pays selon leurs profils d'échanges (**commerce international**) ; (2) regrouper des espèces ou des gènes selon leur similarité.
- **Données et prérequis :** jeu de données de taille modérée, variables standardisées, choix d'un critère de liaison.
- **Peu approprié :** plusieurs dizaines de milliers de points (matrice de distances trop lourde).
- **Métriques :** silhouette, corrélation cophénétique, indice de Rand ajusté si étiquettes réelles.

## 19. DBSCAN

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** DBSCAN (*Density-Based Spatial Clustering of Applications with Noise*). **Famille :** clustering par densité. **Objectif :** trouver des groupes de forme quelconque et isoler le bruit.
- **Intuition :** un groupe est une zone dense de points ; les points isolés sont du bruit.
- **Étapes :** pour chaque point, compter les voisins à distance $\le\varepsilon$ → s'il y en a au moins `min_samples`, c'est un point-cœur → relier les cœurs proches en un même groupe, rattacher les points-frontières, laisser le reste en bruit (étiquette −1).
- **Formule :** $N_\varepsilon(x)=\{x' : d(x,x')\le\varepsilon\}$ ; $x$ est un cœur si $|N_\varepsilon(x)|\ge \text{min\_samples}$.
- **Paramètres :** aucun appris. **Hyperparamètres :** `eps` ($\varepsilon$, ici 0,8) et `min_samples` (ici 5).
- **Avantages :** (1) pas besoin de fixer le nombre de groupes ; (2) détecte des formes arbitraires et les valeurs aberrantes.
- **Limites :** (1) très sensible à `eps` et à des densités différentes selon les groupes ; (2) peine en grande dimension.

### Schéma explicatif

**Comment lire ce schéma.** Vert : points-cœurs (au moins 5 voisins dans le cercle de rayon `eps`) ; jaune : points-frontières ; croix rouges : bruit. Le cercle pointillé montre la zone de voisinage d'un point-cœur.

```python
from sklearn.cluster import DBSCAN
m=DBSCAN(eps=.8,min_samples=5).fit(X_clu);lab=m.labels_;coeur=np.zeros(len(X_clu),bool);coeur[m.core_sample_indices_]=True
fig,ax=plt.subplots(figsize=(5.8,4.4));ax.scatter(*X_clu[coeur].T,s=16,c='tab:green',label='points-cœurs');ax.scatter(*X_clu[~coeur&(lab>=0)].T,s=22,c='gold',edgecolor='k',label='points-frontières')
ax.scatter(*X_clu[lab==-1].T,s=40,c='red',marker='x',label='bruit (étiquette −1)');ax.add_patch(plt.Circle(X_clu[m.core_sample_indices_[0]],.8,fill=False,ls='--'));ax.set_aspect('equal')
ax.set_title('DBSCAN (eps=0,8 ; min_samples=5) : le cercle pointillé a pour rayon eps');ax.legend(fontsize=8);plt.show()
```

### Contextes d'utilisation

- **Cas d'usage :** (1) repérer des zones de forte concentration de commandes ou de livraisons sur une carte (**logistique**) ; (2) isoler des transactions atypiques parmi des groupes de comportements normaux (**finance**).
- **Données et prérequis :** variables standardisées, peu de dimensions, densité homogène.
- **Peu approprié :** groupes de densités très différentes ou données de grande dimension.
- **Métriques :** silhouette (hors bruit), proportion de bruit, stabilité ; indice de Rand ajusté si étiquettes réelles.

```python
from sklearn.cluster import DBSCAN
etiquettes=DBSCAN(eps=.8,min_samples=5).fit_predict(X_clu)
print('Groupes:',set(etiquettes),'Bruit:',(etiquettes==-1).sum())
```

## 20. Mélanges gaussiens

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Mélange gaussien (*Gaussian Mixture Model*, GMM). **Famille :** clustering probabiliste. **Objectif :** modéliser les données comme un mélange de plusieurs lois normales et en déduire des groupes « souples ».
- **Intuition :** chaque groupe est une cloche (gaussienne) qui peut être étirée ou inclinée ; chaque point a une probabilité d'appartenir à chacun.
- **Étapes :** initialiser les cloches → algorithme EM : (E) calculer la probabilité d'appartenance de chaque point à chaque cloche, (M) réajuster moyennes, covariances et poids → répéter jusqu'à convergence.
- **Formule :** $p(x)=\sum_{k=1}^{K}\pi_k\,\mathcal N(x;\mu_k,\Sigma_k)$ ; $\pi_k$ = poids du groupe $k$ (somme égale à 1), $\mu_k$ = moyenne, $\Sigma_k$ = matrice de covariance.
- **Paramètres :** $\pi_k,\mu_k,\Sigma_k$. **Hyperparamètres :** nombre de composantes (ici 3), type de covariance (`full` par défaut), initialisation.
- **Avantages :** (1) groupes de formes elliptiques et probabilités d'appartenance ; (2) critères (BIC, AIC) pour choisir le nombre de groupes.
- **Limites :** (1) suppose des groupes gaussiens ; (2) sensible à l'initialisation, risque de covariances dégénérées.

### Schéma explicatif

**Comment lire ce schéma.** Les ellipses sont les lois normales apprises (contours à 1 et 2 écarts-types) : leur forme et leur inclinaison peuvent différer d'un groupe à l'autre. La couleur d'un point indique sa probabilité d'appartenir au groupe 1 (rouge) ou à l'autre (bleu) ; les points intermédiaires sont incertains.

### Contextes d'utilisation

- **Cas d'usage :** (1) segmenter des clients qui peuvent appartenir à plusieurs profils, avec une probabilité par profil (**marketing**) ; (2) modéliser la distribution de rendements financiers en régimes (calme, crise) (**finance**).

## 21. ACP

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Analyse en composantes principales (*Principal Component Analysis*, ACP / PCA). **Famille :** réduction de dimension (non supervisée). **Objectif :** résumer beaucoup de variables corrélées par quelques axes.
- **Intuition :** on cherche les directions dans lesquelles les données varient le plus et on projette dessus.
- **Étapes :** standardiser → calculer la matrice de covariance → extraire ses vecteurs propres (axes) et valeurs propres (variances) → projeter sur les $q$ premiers axes.
- **Formule :** composante $k$ : $z_k=w_k^\top x$ où $w_k$ est le vecteur propre associé à la $k$-ième plus grande valeur propre $\lambda_k$ ; part de variance expliquée $\lambda_k/\sum_j\lambda_j$.
- **Paramètres :** les axes $w_k$. **Hyperparamètre :** nombre de composantes $q$ (ici 2).
- **Avantages :** (1) réduit le bruit et la redondance ; (2) permet de visualiser des données en 2D et résultat reproductible.
- **Limites :** (1) linéaire seulement ; (2) les composantes sont des combinaisons de variables, difficiles à interpréter, et la méthode est sensible à l'échelle.

### Schéma explicatif

**Comment lire ce schéma.** Les points gris sont les données centrées. La flèche rouge est la 1re composante (direction de plus grande dispersion), la bleue la 2e, perpendiculaire. Projeter sur la flèche rouge conserve la plus grande part de la variance.

```python
rng=np.random.default_rng(GRAINE);P=rng.normal(size=(120,2))@np.array([[2.2,1.2],[0,.6]]);P-=P.mean(0)
vp,V=np.linalg.eigh(np.cov(P.T));V=V[:,::-1];vp=vp[::-1];fig,ax=plt.subplots(figsize=(5.2,4.2));ax.scatter(P[:,0],P[:,1],s=12,c='gray',label='données centrées')
for k,c in enumerate(['red','blue']): ax.annotate('',xy=V[:,k]*3*np.sqrt(vp[k]),xytext=(0,0),arrowprops=dict(arrowstyle='->',lw=2.5,color=c));ax.plot([],[],c=c,lw=2.5,label=f'composante {k+1} ({vp[k]/vp.sum():.0%} de la variance)')
ax.set_aspect('equal');ax.set_title('ACP : axes orthogonaux de variance maximale');ax.legend(fontsize=8);plt.show()
```

### Contextes d'utilisation

- **Cas d'usage :** (1) condenser des dizaines d'indicateurs économiques de pays en 2 ou 3 axes (**commerce international**) ; (2) réduire des centaines de rendements d'actions en quelques facteurs de risque (**finance**).
- **Données et prérequis :** variables numériques corrélées et standardisées.
- **Peu approprié :** structures non linéaires (spirale, anneau) : l'ACP ne les « déplie » pas.
- **Métriques :** variance expliquée (cumulée), erreur de reconstruction.

```python
from sklearn.decomposition import PCA
X_vin=StandardScaler().fit_transform(load_wine().data)
modele=PCA(n_components=2).fit(X_vin)
projection=modele.transform(X_vin)
print('Variance expliquée:',modele.explained_variance_ratio_)
plt.scatter(projection[:,0],projection[:,1],s=10);plt.show()
```

- **Sortie :** variance expliquée = **[0,362 ; 0,192]** (soit environ 36,2 % puis 19,2 %, au total environ 55 %) et un nuage 2D.

## 22. t-SNE

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** t-SNE (*t-distributed Stochastic Neighbor Embedding*). **Famille :** réduction de dimension non linéaire, surtout pour la visualisation. **Objectif :** représenter en 2D des données de grande dimension en conservant les voisinages.
- **Intuition :** des points voisins dans l'espace d'origine doivent rester voisins sur la carte 2D.
- **Étapes :** calculer des probabilités de voisinage $p_{ij}$ dans l'espace d'origine (noyau gaussien) → placer les points au hasard en 2D → calculer des probabilités $q_{ij}$ avec un noyau de Student → déplacer les points par descente de gradient pour rapprocher $q$ de $p$.
- **Formule :** minimiser $\mathrm{KL}(P\|Q)=\sum_{i\ne j}p_{ij}\ln\frac{p_{ij}}{q_{ij}}$ ; $p_{ij}$ et $q_{ij}$ = similarités entre $i$ et $j$ avant et après projection.
- **Paramètres :** les coordonnées 2D. **Hyperparamètres :** `perplexity` (ici 20, voisinage effectif), nombre d'itérations, taux d'apprentissage.
- **Avantages :** (1) révèle bien les groupes locaux ; (2) très utilisé pour explorer des données complexes.
- **Limites :** (1) les distances entre groupes et la taille des groupes n'ont pas de sens, et le résultat varie avec la graine ; (2) lent sur de grands jeux de données et pas de transformation de nouvelles données.

### Schéma explicatif

**Comment lire ce schéma.** À gauche : le noyau de Student a des queues plus épaisses que le noyau gaussien, ce qui permet de séparer les groupes en 2D sans les tasser. À droite : projection t-SNE des 13 mesures des vins, colorée par cépage (couleur connue mais non utilisée pour projeter).

```python
from sklearn.datasets import load_wine
from sklearn.manifold import TSNE
fig,axs=plt.subplots(1,2,figsize=(10,3.8));t=np.linspace(-6,6,300);axs[0].plot(t,np.exp(-t**2/2),label='noyau gaussien (espace d’origine)');axs[0].plot(t,1/(1+t**2),label='noyau de Student t (espace 2D)')
axs[0].set_title('Queues plus épaisses en 2D : évite l’entassement',fontsize=9);axs[0].set_xlabel('distance');axs[0].set_ylabel('similarité');axs[0].legend(fontsize=8)
V=load_wine();Xv=StandardScaler().fit_transform(V.data);E=TSNE(perplexity=20,random_state=GRAINE).fit_transform(Xv)
axs[1].scatter(E[:,0],E[:,1],c=V.target,cmap='viridis',s=12);axs[1].set_title('t-SNE sur 13 variables (couleur = cépage connu)',fontsize=9);plt.show()
```

### Contextes d'utilisation

- **Cas d'usage :** (1) explorer visuellement des représentations de produits ou de clients (plongements) pour repérer des segments (**marketing**) ; (2) visualiser des cellules en biologie à partir de milliers de gènes.
- **Données et prérequis :** données standardisées, jusqu'à quelques dizaines de milliers de points ; essayer plusieurs valeurs de `perplexity`.
- **Peu approprié :** comme étape de prétraitement pour un modèle prédictif ou pour mesurer des distances entre groupes.
- **Métriques :** *trustworthiness* (conservation des voisinages), divergence KL ; interprétation visuelle prudente.

```python
from sklearn.manifold import TSNE
X_vin=StandardScaler().fit_transform(load_wine().data)
projection=TSNE(n_components=2,perplexity=20,random_state=GRAINE).fit_transform(X_vin)
plt.scatter(projection[:,0],projection[:,1],s=10);plt.title('t-SNE');plt.show()
```

## 23. UMAP

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** UMAP (*Uniform Manifold Approximation and Projection*). **Famille :** réduction de dimension non linéaire. **Objectif :** projeter en 2D (ou plus) en conservant la structure locale et en partie la structure globale.
- **Intuition :** on construit un graphe de voisinage des données, puis on cherche une disposition en basse dimension qui lui ressemble.
- **Étapes :** graphe des plus proches voisins → pondération « floue » des liens → disposition initiale → optimisation par descente de gradient (attraction des voisins, répulsion des autres).
- **Formule :** minimiser l'entropie croisée entre les graphes flous d'origine ($\mu_{ij}$) et de projection ($\nu_{ij}$) : $\sum_{i\ne j}\big[\mu_{ij}\ln\frac{\mu_{ij}}{\nu_{ij}}+(1-\mu_{ij})\ln\frac{1-\mu_{ij}}{1-\nu_{ij}}\big]$.
- **Paramètres :** les coordonnées projetées. **Hyperparamètres :** `n_neighbors` (ici 15), `min_dist` (écart minimal entre points), dimension de sortie.
- **Avantages :** (1) plus rapide que t-SNE et utilisable sur de nouvelles données ; (2) préserve mieux la structure globale.
- **Limites :** (1) dépend de bibliothèque externe et de réglages ; (2) les distances restent difficiles à interpréter et le résultat varie avec la graine.

### Schéma explicatif

**Comment lire ce schéma.** À gauche : étape 1 d'UMAP, le graphe où chaque point est relié à ses 3 plus proches voisins. À droite : les 4 étapes de l'algorithme, de haut en bas.

```python
from sklearn.neighbors import NearestNeighbors
Xs=X_clu[np.random.default_rng(GRAINE).choice(len(X_clu),40,replace=False)];nn=NearestNeighbors(n_neighbors=4).fit(Xs).kneighbors(Xs)[1]
fig,axs=plt.subplots(1,2,figsize=(10.5,4),gridspec_kw={'width_ratios':[1.2,1]});ax=axs[0]
for i in range(len(Xs)):
    for j in nn[i,1:]: ax.plot(*zip(Xs[i],Xs[j]),c='gray',lw=.6)
ax.scatter(Xs[:,0],Xs[:,1],s=22,c='tab:blue',zorder=3);ax.set_title('Étape 1 : graphe des 3 plus proches voisins',fontsize=9)
flux(['1. Graphe de voisinage (n_neighbors) pondéré','2. Représentation « floue » des liens (proba. de voisinage)','3. Disposition initiale en 2D (spectrale)','4. Optimisation : rapprocher les voisins, écarter les autres\n(contrôlé par min_dist)'],'Les 4 étapes d’UMAP',ax=axs[1]);plt.show()
```

### Contextes d'utilisation

- **Cas d'usage :** (1) cartographier des milliers de clients selon leur comportement pour définir des segments marketing (**marketing**) ; (2) visualiser des plongements de textes ou de produits.
- **Données et prérequis :** données standardisées, bibliothèque `umap-learn`.
- **Peu approprié :** quand l'interprétabilité des axes est nécessaire (préférer l'ACP).

## 24. Isolation Forest

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Isolation Forest (forêt d'isolement). **Famille :** détection d'anomalies (non supervisée). **Objectif :** repérer les observations atypiques.
- **Intuition :** un point aberrant est isolé en très peu de coupures aléatoires, alors qu'un point ordinaire exige beaucoup de coupures.
- **Étapes :** construire de nombreux arbres en choisissant variable et seuil au hasard → mesurer la profondeur moyenne à laquelle chaque point est isolé → convertir en score d'anomalie → signaler les scores les plus extrêmes.
- **Formule :** $s(x,n)=2^{-E[h(x)]/c(n)}$ ; $h(x)$ = longueur du chemin dans un arbre, $E[h(x)]$ = moyenne sur la forêt, $c(n)$ = longueur moyenne de référence pour $n$ points ; $s$ proche de 1 = anomalie.
- **Paramètres :** les arbres. **Hyperparamètres :** `n_estimators`, `contamination` (proportion attendue d'anomalies, ici 0,02), taille des sous-échantillons.
- **Avantages :** (1) rapide, adapté aux grandes données ; (2) sans hypothèse sur la forme des données normales.
- **Limites :** (1) la proportion `contamination` doit être choisie et conditionne le nombre d'alertes ; (2) anomalies « cachées » au sein d'un amas difficiles à voir.

### Schéma explicatif

**Comment lire ce schéma.** La couleur est le score d'anomalie : vert = normal, rouge = suspect (score bas). Les cercles rouges sont les points signalés avec `contamination=0.02`. Les points isolés loin des nuages sont les plus rouges.

```python
from sklearn.ensemble import IsolationForest
Xa=np.vstack([X_clu,[[9,9],[-8,-8]]]);m=IsolationForest(contamination=.02,random_state=GRAINE).fit(Xa);s=m.score_samples(Xa);flag=m.predict(Xa)==-1
fig,ax=plt.subplots(figsize=(5.8,4.4));sc=ax.scatter(Xa[:,0],Xa[:,1],c=s,cmap='RdYlGn',s=22);plt.colorbar(sc,label='score d’anomalie (bas = suspect)')
ax.scatter(Xa[flag,0],Xa[flag,1],s=150,facecolors='none',edgecolors='red',lw=2,label='signalés (contamination=2 %)');ax.set_title('Isolation Forest : les points faciles à isoler ont un score bas');ax.legend(fontsize=8);plt.show()
```

- **Données et prérequis :** variables numériques, très peu d'anomalies, pas besoin d'étiquettes.
- **Peu approprié :** quand les anomalies sont nombreuses (plus de 10 à 20 %) ou quand elles se situent à l'intérieur de la zone normale (anomalies contextuelles).

## 25. Bandits manchots

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Bandit manchot (*Multi-armed bandit*). **Famille :** apprentissage par renforcement (cas à un seul état). **Objectif :** choisir à répétition parmi plusieurs options (« bras ») pour maximiser le gain cumulé alors que leur rendement est inconnu.
- **Intuition :** il faut tester les bras pour les connaître (exploration) mais aussi jouer le meilleur connu (exploitation).
- **Étapes :** à chaque tour, choisir un bras → observer la récompense → mettre à jour l'estimation de valeur du bras. Ici le code joue les bras à tour de rôle : c'est l'exploration pure, donc une base de comparaison.
- **Formule :** estimation de valeur du bras $a$ : $\hat Q_a=\frac{1}{N_a}\sum_{t:a_t=a}r_t$ ; regret cumulé $R_T=\sum_{t=1}^{T}\big(\mu^\star-\mu_{a_t}\big)$ ; $N_a$ = nombre de tirages du bras $a$, $\mu^\star$ = valeur du meilleur bras.
- **Paramètres :** les estimations $\hat Q_a$. **Hyperparamètres :** règle de choix du bras (rotation, $\varepsilon$-greedy, UCB, Thompson), nombre de tours.
- **Avantages :** (1) cadre simple pour décider sous incertitude ; (2) apprentissage continu pendant l'utilisation.
- **Limites :** (1) ne tient pas compte du contexte ni des états successifs ; (2) suppose des bras stables dans le temps.

### Schéma explicatif

**Comment lire ce schéma.** Barres bleues : vraies probabilités de gain de chaque bras (inconnues de l'agent). Barres orange : probabilités estimées après 500 tirages joués en rotation. Plus elles sont proches, meilleure est l'estimation.

```python
rng=np.random.default_rng(GRAINE);vrai=np.array([.2,.5,.7]);gains=np.zeros(3);n=np.zeros(3)
for t in range(500):
    b=t%3;gains[b]+=rng.random()<vrai[b];n[b]+=1
fig,ax=plt.subplots(figsize=(5.2,3.6));pos=np.arange(3);ax.bar(pos-.2,vrai,.4,label='probabilité de gain réelle (inconnue de l’agent)');ax.bar(pos+.2,gains/n,.4,label='estimation après 500 tirages en rotation')
ax.set_xticks(pos);ax.set_xticklabels(['bras 0','bras 1','bras 2']);ax.set_ylabel('probabilité de gain');ax.set_title('Bandit à 3 bras : estimer la valeur de chaque bras');ax.legend(fontsize=8);plt.show()
```

### Contextes d'utilisation

- **Cas d'usage :** (1) test de plusieurs versions d'une publicité ou d'un prix pour maximiser le taux de clic (**marketing**) ; (2) choix du meilleur transporteur parmi plusieurs pour un trajet (**logistique**).
- **Données et prérequis :** retour de récompense immédiat après chaque choix, nombre de bras limité.
- **Peu approprié :** décisions dont la récompense dépend d'une suite d'actions (utiliser le Q-learning) ou environnement changeant vite.
- **Métriques :** récompense cumulée, regret cumulé, part de choix du meilleur bras, variation entre graines.

```python
generateur=np.random.default_rng(GRAINE)
vraies_probas=np.array([.2,.5,.7]);somme=np.zeros(3);comptes=np.zeros(3)
for t in range(500):
    bras=t%3
    gain=int(generateur.random()<vraies_probas[bras]);somme[bras]+=gain;comptes[bras]+=1
print('Valeurs estimées:',somme/comptes)
```

### Analyse de l'expérience Python

- **Entrées :** 3 bras dont les vraies probabilités de gain sont 0,2 ; 0,5 et 0,7 (inconnues de l'agent), 500 tirages en rotation (graine 42). **Hyperparamètres :** aucun apprentissage de choix.
- **Sortie :** valeurs estimées = **[0,180 ; 0,539 ; 0,723]**.
- **Lecture :** les estimations sont proches des vraies valeurs (0,2 ; 0,5 ; 0,7) parce que chaque bras est joué environ 167 fois.
- **Limite :** cette stratégie ne cherche pas à gagner : elle joue autant les mauvais bras que le bon. C'est une référence d'exploration pure, pas un algorithme optimisé. Ce sont des données simulées.

## 26. Epsilon-greedy

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Epsilon-greedy (*ε-greedy*). **Famille :** apprentissage par renforcement (stratégie pour les bandits). **Objectif :** équilibrer exploration et exploitation par une règle très simple.
- **Intuition :** la plupart du temps on joue le meilleur bras connu, mais avec une petite probabilité $\varepsilon$ on en joue un au hasard pour continuer à explorer.
- **Étapes :** à chaque tour : tirer un nombre au hasard → s'il est inférieur à $\varepsilon$, choisir un bras au hasard, sinon le bras de plus forte valeur estimée → observer la récompense → mettre à jour les compteurs.
- **Formule :** $a_t=\begin{cases}\text{bras au hasard} & \text{avec probabilité }\varepsilon\\ \arg\max_a \hat Q_a & \text{avec probabilité }1-\varepsilon\end{cases}$ ; $\hat Q_a$ = valeur estimée du bras $a$ (ici lissée par $(S_a+1)/(N_a+1)$, ce qui donne un biais optimiste au départ).
- **Paramètres :** les estimations $\hat Q_a$. **Hyperparamètre :** $\varepsilon$ (ici 0,1), éventuellement décroissant.
- **Avantages :** (1) très simple à coder ; (2) garantit que tous les bras continuent à être testés.
- **Limites :** (1) explore au hasard même quand tout est connu (regret qui augmente sans cesse) ; (2) le choix de $\varepsilon$ est arbitraire.

### Schéma explicatif

**Comment lire ce schéma.** L'axe vertical est le **regret cumulé** (gain perdu par rapport au meilleur bras) : plus la courbe est basse, mieux c'est. ε=1 (tout au hasard) perd constamment ; ε=0,1 apprend puis exploite ; ε=0 n'explore jamais et peut rester bloqué sur un mauvais bras.

```python
def regret(eps,T=1000,graine=GRAINE):
    r=np.random.default_rng(graine);vrai=np.array([.2,.5,.7]);S=np.zeros(3);N=np.zeros(3);cum=[];tot=0
    for t in range(T):
        b=int(r.integers(3)) if r.random()<eps else int(np.argmax((S+1)/(N+1)));S[b]+=r.random()<vrai[b];N[b]+=1;tot+=vrai.max()-vrai[b];cum.append(tot)
    return cum
fig,ax=plt.subplots(figsize=(5.6,3.6))
for e,c in [(1.0,'tab:red'),(.1,'tab:green'),(.0,'tab:blue')]: ax.plot(regret(e),c=c,label=f'ε = {e}')
ax.set_xlabel('nombre de tirages');ax.set_ylabel('regret cumulé');ax.set_title('Epsilon-greedy : compromis exploration / exploitation');ax.legend(fontsize=8);plt.show()
```

### Contextes d'utilisation

- **Cas d'usage :** (1) allocation du trafic entre plusieurs pages d'un site marchand (**commerce en ligne**) ; (2) sélection d'un fournisseur de livraison parmi plusieurs (**logistique**).
- **Données et prérequis :** récompense observable à chaque tour, peu de bras.
- **Peu approprié :** coût d'une mauvaise action très élevé (l'exploration aléatoire est dangereuse) ou récompenses qui changent selon un contexte (utiliser des bandits contextuels).
- **Métriques :** récompense cumulée, regret, fréquence de choix du meilleur bras, variation entre graines.

```python
generateur=np.random.default_rng(GRAINE)
vraies_probas=np.array([.2,.5,.7]);somme=np.zeros(3);comptes=np.zeros(3);epsilon=.1
for t in range(1000):
    bras=int(generateur.integers(3)) if generateur.random()<epsilon else int(np.argmax((somme+1)/(comptes+1)))
    gain=int(generateur.random()<vraies_probas[bras]);somme[bras]+=gain;comptes[bras]+=1
print('Nombre de choix:',comptes,'Valeurs estimées:',somme/np.maximum(comptes,1))
```

### Analyse de l'expérience Python

- **Entrées :** mêmes 3 bras (0,2 ; 0,5 ; 0,7), 1000 tours. **Hyperparamètre :** `epsilon=0.1`.
- **Sortie :** nombre de choix = **[25, 49, 926]** ; valeurs estimées = **[0,200 ; 0,673 ; 0,693]**.
- **Lecture :** l'agent a joué 926 fois le meilleur bras (le bras 2). L'estimation du bras 1 (0,673) est bien au-dessus de sa vraie valeur (0,5) : avec seulement 49 tirages, le hasard de l'échantillonnage pèse beaucoup.
- **Limite :** une seule graine est utilisée ; il faudrait répéter avec plusieurs graines pour mesurer la variabilité. Données simulées.

## 27. Q-learning

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Q-learning. **Famille :** apprentissage par renforcement (méthode tabulaire, « hors politique »). **Objectif :** apprendre quelle action choisir dans chaque état pour maximiser la récompense cumulée.
- **Intuition :** on tient une table $Q(s,a)$ qui estime la valeur de l'action $a$ dans l'état $s$ et on la corrige à chaque expérience.
- **Étapes :** observer l'état $s$ → choisir une action (ε-greedy) → recevoir la récompense $r$ et le nouvel état $s'$ → mettre à jour $Q(s,a)$ → répéter sur de nombreux épisodes.
- **Formule :** $Q(s,a)\leftarrow Q(s,a)+\alpha\big[r+\gamma\max_{a'}Q(s',a')-Q(s,a)\big]$ ; $\alpha$ = taux d'apprentissage (ici 0,3), $\gamma$ = facteur d'actualisation (ici 0,95), $s'$ = état suivant.
- **Paramètres :** la table $Q$. **Hyperparamètres :** $\alpha$, $\gamma$, $\varepsilon$ (ici décroissant de 1 à 0,05), nombre d'épisodes (ici 4000).
- **Avantages :** (1) converge vers la politique optimale dans le cas tabulaire ; (2) apprend à partir d'actions exploratoires.
- **Limites :** (1) impraticable si le nombre d'états est très grand ; (2) peut apprendre des politiques risquées pendant l'exploration.

### Schéma explicatif

**Comment lire ce schéma.** L'agent choisit une action, l'environnement répond par un nouvel état et une récompense. La formule en jaune met à jour $Q(s,a)$ : la cible utilise $\max_{a'}Q(s',a')$, c'est-à-dire la meilleure action suivante, même si l'agent n'a pas joué celle-là.

```python
boucle('Q-learning : apprentissage « off-policy »','$Q(s,a)\\leftarrow Q(s,a)+\\alpha\\,[\\,r+\\gamma\\,\\max_{a\'}Q(s\',a\')-Q(s,a)\\,]$','la cible utilise la MEILLEURE action suivante, même si l’agent ne la joue pas');plt.show()
```

### Contextes d'utilisation

- **Cas d'usage :** (1) tarification dynamique ou gestion de stocks avec un petit nombre d'états (**gestion / commerce**) ; (2) navigation d'un robot d'entrepôt sur une grille (**logistique**).
- **Données et prérequis :** environnement simulable (états et actions discrets), récompense définie.
- **Peu approprié :** espaces d'états continus ou immenses (images) : utiliser le DQN ; situations où l'on ne peut pas simuler d'essais.
- **Métriques :** récompense cumulée, taux de réussite, nombre de pas jusqu'au but, variation entre graines.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False)
Q=np.zeros((env.observation_space.n,env.action_space.n));generateur=np.random.default_rng(GRAINE)
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode)
    for pas in range(100):
        action=int(generateur.integers(env.action_space.n)) if generateur.random()<max(.05,1-episode/2500) else int(np.argmax(Q[etat]))
        suivant,recompense,termine,tronque,_=env.step(action)
        cible=recompense+(0 if termine else .95*np.max(Q[suivant]))
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat=suivant
        if termine or tronque:break
print('Politique:',Q.argmax(axis=1));env.close()
```

### Analyse de l'expérience Python

- **Entrées :** environnement Gymnasium `FrozenLake-v1` (grille 4×4, sans glissement, `is_slippery=False`). **Hyperparamètres :** 4000 épisodes, $\alpha=0{,}3$, $\gamma=0{,}95$, $\varepsilon$ décroissant.
- **Sortie :** *à compléter après exécution sur Colab* : la politique affichée est le vecteur de 16 actions (0 = gauche, 1 = bas, 2 = droite, 3 = haut). La cellule n'a pas pu être exécutée lors de la préparation de ce notebook, car `gymnasium` n'y était pas installé.
- **Lecture attendue :** en partant de l'état 0, suivre les actions de la politique doit mener à l'état 15 (le but) en 6 pas. Les états jamais visités gardent l'action 0 par défaut (valeurs Q nulles).
- **Limite :** une seule graine ; le glissement est désactivé, ce qui rend le problème déterministe et facile.

## 28. SARSA

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** SARSA (*State–Action–Reward–State–Action*). **Famille :** apprentissage par renforcement (méthode tabulaire, « sur politique »). **Objectif :** apprendre une table $Q(s,a)$ en tenant compte de la façon dont l'agent explore réellement.
- **Intuition :** comme le Q-learning, mais la mise à jour utilise l'action que l'agent va effectivement jouer ensuite, pas la meilleure action possible.
- **Étapes :** choisir $a$ (ε-greedy) → recevoir $r,s'$ → choisir déjà $a'$ (ε-greedy) → mettre à jour $Q(s,a)$ avec $Q(s',a')$ → continuer avec $(s',a')$.
- **Formule :** $Q(s,a)\leftarrow Q(s,a)+\alpha\big[r+\gamma\,Q(s',a')-Q(s,a)\big]$ ; mêmes symboles que le Q-learning, $a'$ = action réellement choisie dans $s'$.
- **Paramètres :** la table $Q$. **Hyperparamètres :** $\alpha$ (0,3), $\gamma$ (0,95), $\varepsilon$ décroissant, nombre d'épisodes (4000).
- **Avantages :** (1) apprentissage plus prudent près des zones dangereuses ; (2) s'appuie sur la politique effectivement suivie.
- **Limites :** (1) converge vers la politique optimale seulement si $\varepsilon$ tend vers 0 ; (2) mêmes limites que le Q-learning pour les grands espaces d'états.

### Schéma explicatif

**Comment lire ce schéma.** Même boucle que pour le Q-learning ; seule la cible change : elle utilise $Q(s',a')$ de l'action $a'$ que l'agent a **réellement** choisie (donc avec son exploration). L'apprentissage tient compte de ses propres risques.

```python
boucle('SARSA : apprentissage « on-policy »','$Q(s,a)\\leftarrow Q(s,a)+\\alpha\\,[\\,r+\\gamma\\,Q(s\',a\')-Q(s,a)\\,]$','la cible utilise l’action $a\'$ réellement choisie (avec exploration ε)');plt.show()
```

### Contextes d'utilisation

- **Cas d'usage :** (1) robot ou véhicule autonome devant éviter une zone dangereuse pendant l'apprentissage (**logistique**) ; (2) gestion de stock où une exploration trop risquée a un coût réel (**gestion**).
- **Données et prérequis :** environnement simulable avec états et actions discrets.
- **Peu approprié :** espaces d'états continus ; si l'on veut la politique optimale « gloutonne » sans tenir compte de l'exploration, le Q-learning convient mieux.
- **Métriques :** récompense cumulée pendant l'apprentissage, taux de réussite, variation entre graines, comparaison avec Q-learning.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False);Q=np.zeros((16,4));generateur=np.random.default_rng(GRAINE)
def choisir(etat,epsilon):
    return int(generateur.integers(4)) if generateur.random()<epsilon else int(np.argmax(Q[etat]))
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode);epsilon=max(.05,1-episode/2500);action=choisir(etat,epsilon)
    for pas in range(100):
        suivant,recompense,termine,tronque,_=env.step(action)
        action_suiv=choisir(suivant,epsilon)
        cible=recompense+(0 if termine else .95*Q[suivant,action_suiv])
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat,action=suivant,action_suiv
        if termine or tronque:break
print('Politique SARSA:',Q.argmax(axis=1));env.close()
```

### Analyse de l'expérience Python

- **Entrées :** même environnement `FrozenLake-v1` sans glissement. **Hyperparamètres :** identiques au Q-learning.
- **Sortie :** *à compléter après exécution sur Colab* : vecteur de 16 actions. Cellule non exécutée lors de la préparation (module `gymnasium` absent).
- **Lecture attendue :** comparer cette politique à celle du Q-learning (exemple 27) ; sur une grille sans danger de glissement, elles devraient être proches. La différence se voit surtout dans des environnements risqués (falaise).
- **Limite :** une seule graine ; pour juger, comparer plusieurs graines et mesurer le taux de réussite (partie B).

## 29. DQN — cible conceptuelle

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Réseau Q profond (*Deep Q-Network*, DQN). **Famille :** apprentissage par renforcement profond. **Objectif :** appliquer le Q-learning quand les états sont trop nombreux ou continus, en approchant $Q(s,a)$ par un réseau de neurones.
- **Intuition :** au lieu d'une table, un réseau de paramètres $\theta$ prédit la valeur de chaque action à partir de l'état.
- **Étapes :** jouer et stocker les transitions $(s,a,r,s')$ dans une mémoire de rejeu → tirer un mini-lot au hasard → calculer la cible avec un réseau « cible » figé → ajuster le réseau en ligne par descente de gradient → recopier périodiquement les paramètres dans le réseau cible.
- **Formule :** cible $y=r+\gamma(1-\text{fin})\max_{a'}Q(s',a';\theta^-)$ et perte $(y-Q(s,a;\theta))^2$ ; $\theta$ = paramètres du réseau en ligne, $\theta^-$ = paramètres du réseau cible, « fin » vaut 1 si l'épisode est terminé.
- **Paramètres :** les poids $\theta$. **Hyperparamètres :** $\gamma$, taille de la mémoire, taille du mini-lot, fréquence de recopie $C$, architecture, $\varepsilon$.
- **Avantages :** (1) gère des espaces d'états énormes (images, signaux) ; (2) la mémoire de rejeu réutilise les expériences.
- **Limites :** (1) entraînement instable et gourmand en calcul ; (2) sensible aux hyperparamètres et peu interprétable.

### Schéma explicatif

**Comment lire ce schéma.** Lire de haut en bas : l'agent joue et stocke ses transitions dans la mémoire de rejeu ; on tire un mini-lot au hasard, on calcule la cible avec un réseau figé $\theta^-$, on ajuste le réseau en ligne $\theta$ pour réduire l'écart, et on recopie périodiquement $\theta$ vers $\theta^-$.

```python
flux(['Environnement → transition $(s,a,r,s\')$','Mémoire de rejeu : stockage des transitions passées','Mini-lot tiré AU HASARD dans la mémoire (casse les corrélations)','Cible $y=r+\\gamma(1-\\mathrm{fin})\\max_{a\'}Q(s\',a\';\\theta^-)$  (réseau cible figé)','Perte $(y-Q(s,a;\\theta))^2$ → descente de gradient sur le réseau en ligne $\\theta$','Tous les C pas : copier $\\theta^-\\leftarrow\\theta$'],'DQN : Q-learning avec réseau de neurones, rejeu et réseau cible');plt.show()
```

### Contextes d'utilisation

- **Cas d'usage :** (1) optimisation dynamique de la gestion d'entrepôt ou de flottes avec un grand nombre d'états (**logistique**) ; (2) agents de jeux vidéo (Atari) ou pilotage de systèmes énergétiques.
- **Données et prérequis :** simulateur ou environnement où l'on peut jouer un très grand nombre d'épisodes, bibliothèque de réseaux de neurones (PyTorch, TensorFlow).
- **Peu approprié :** petits espaces d'états (une table $Q$ suffit) ou situations où les essais coûtent cher dans le monde réel.
- **Métriques :** récompense cumulée moyenne, taux de réussite, stabilité entre graines, perte d'apprentissage.

### Analyse de l'expérience Python

- **Entrées :** trois transitions fictives : récompenses $[0,1,-1]$, indicateur de fin $[\text{non},\text{oui},\text{non}]$, meilleure valeur suivante $[2,3,1]$, $\gamma=0{,}95$.
- **Sortie :** cibles de Bellman = **[1,9 ; 1,0 ; −0,05]**.
- **Lecture :** transition 1 : $0+0{,}95\times2=1{,}9$ ; transition 2 : épisode terminé, la cible vaut simplement la récompense (1) ; transition 3 : $-1+0{,}95\times1=-0{,}05$.
- **Limite :** c'est un calcul conceptuel, aucun réseau n'est entraîné ; il ne dit rien des performances d'un DQN complet (expérience facultative selon la consigne).

## Q7 / Q8 — Clustering et réduction de dimension sur données réelles (Wine)

Les 13 mesures sont standardisées (sans cible, pas de fuite possible). Le cépage connu n'est utilisé **que pour l'évaluation externe** (indice de Rand ajusté, ARI). **Critères internes :** silhouette. **Stabilité :** ARI moyen entre la partition complète et celles obtenues sur 10 sous-échantillons de 80 %.

```python
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN
from sklearn.mixture import GaussianMixture
from sklearn.metrics import silhouette_score, adjusted_rand_score

Xs_vin = StandardScaler().fit_transform(X_vin); y_ref = y_vin.to_numpy()

# eps de DBSCAN choisi par un critère INTERNE (silhouette hors bruit), sans utiliser les étiquettes
meilleur = None
for eps in np.arange(1.5, 4.6, .25):
    lab = DBSCAN(eps=eps, min_samples=5).fit_predict(Xs_vin); ok = lab >= 0
    if len(set(lab[ok])) >= 2 and ok.mean() >= .7:
        s = silhouette_score(Xs_vin[ok], lab[ok])
        if meilleur is None or s > meilleur[1]: meilleur = (float(eps), s)
eps_db = meilleur[0] if meilleur else 2.5
print('eps retenu pour DBSCAN :', eps_db)

algos = {'K-means (référence)': lambda X: KMeans(n_clusters=3, n_init=10, random_state=GRAINE).fit_predict(X),
         'Hiérarchique (Ward)': lambda X: AgglomerativeClustering(n_clusters=3).fit_predict(X),
         'Mélange gaussien': lambda X: GaussianMixture(n_components=3, random_state=GRAINE).fit(X).predict(X),
         'DBSCAN': lambda X: DBSCAN(eps=eps_db, min_samples=5).fit_predict(X)}
def stabilite(ajuste, X, n=10):
    rng = np.random.default_rng(GRAINE); ref = ajuste(X); v = []
    for _ in range(n):
        idx = rng.choice(len(X), int(.8*len(X)), replace=False); v.append(adjusted_rand_score(ref[idx], ajuste(X[idx])))
    return float(np.mean(v))
lignes = []
for nom, f in algos.items():
    t0 = time.time(); lab = f(Xs_vin); duree = time.time() - t0; ok = lab >= 0
    sil = silhouette_score(Xs_vin[ok], lab[ok]) if len(set(lab[ok])) >= 2 else np.nan
    lignes.append({'méthode': nom, 'groupes': len(set(lab[ok])), 'bruit (%)': 100*(~ok).mean(), 'silhouette': sil,
                   'ARI vs cépages (externe)': adjusted_rand_score(y_ref, lab), 'stabilité (ARI)': stabilite(f, Xs_vin), 'durée (s)': duree})
tab_clu = pd.DataFrame(lignes).set_index('méthode'); RESULTATS['Clustering (Wine)'] = tab_clu; display(tab_clu.round(3))
```

```python
# Réduction de dimension : ACP (référence), t-SNE, UMAP
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE, trustworthiness
cum = np.cumsum(PCA().fit(Xs_vin).explained_variance_ratio_)
print('ACP : variance expliquée par 2 axes =', round(float(cum[1]), 3), '| nombre d\'axes pour 90 % =', int(np.argmax(cum >= .9)) + 1)
proj = {'ACP (2 axes)': PCA(n_components=2).fit_transform(Xs_vin),
        't-SNE (perplexité 20)': TSNE(n_components=2, perplexity=20, random_state=GRAINE).fit_transform(Xs_vin)}
if umap is not None: proj['UMAP (15 voisins)'] = umap.UMAP(n_neighbors=15, random_state=GRAINE).fit_transform(Xs_vin)
tab_red = pd.DataFrame({'trustworthiness (10 voisins)': {k: trustworthiness(Xs_vin, v, n_neighbors=10) for k, v in proj.items()}})
RESULTATS['Réduction de dimension (Wine)'] = tab_red; display(tab_red.round(3))
fig, axs = plt.subplots(1, len(proj), figsize=(4.2*len(proj), 3.6))
for ax, (nom, P) in zip(np.atleast_1d(axs), proj.items()):
    ax.scatter(P[:, 0], P[:, 1], c=y_ref, cmap='viridis', s=14); ax.set_title(nom, fontsize=10)
fig.suptitle('Wine : projections 2D (couleur = cépage connu, non utilisé pour projeter)'); plt.tight_layout(); plt.show()
```

## Q7 / Q8 — Détection d'anomalies (Breast Cancer Wisconsin)

**Construction du jeu :** on garde tous les cas bénins (357, considérés comme « normaux ») et on tire au hasard 20 cas malins (« anomalies », environ 5 %). Ce sous-échantillonnage est une construction classique pour tester des détecteurs d'anomalies avec des étiquettes fiables ; **il modifie la proportion réelle de cas malins**, ce qui doit être signalé comme limite. **Méthodes :** référence simple = plus grand écart (en écarts-types) sur une variable ; méthodes avancées = Isolation Forest et Local Outlier Factor (LOF). **Hypothèse favorable :** on donne aux méthodes la vraie proportion d'anomalies (`contamination`) pour fixer le nombre d'alertes ; la ROC-AUC et la PR-AUC, calculées sur les scores, n'en dépendent pas.

```python
from sklearn.ensemble import IsolationForest
from sklearn.neighbors import LocalOutlierFactor
rng = np.random.default_rng(GRAINE); yc = cancer.target.to_numpy()          # 0 = malin, 1 = bénin
idx = np.r_[np.where(yc == 1)[0], rng.choice(np.where(yc == 0)[0], 20, replace=False)]
X_an = StandardScaler().fit_transform(cancer.data.iloc[idx]); y_an = (yc[idx] == 0).astype(int)     # 1 = anomalie
k = int(y_an.sum()); contam = float(y_an.mean()); print('observations :', len(y_an), '| anomalies :', k, f'({contam:.1%})')

def evalue(nom, score, alertes):
    return {'méthode': nom, 'ROC-AUC': roc_auc_score(y_an, score), 'PR-AUC': average_precision_score(y_an, score),
            'précision (alertes)': precision_score(y_an, alertes), 'rappel (alertes)': recall_score(y_an, alertes), 'nb alertes': int(alertes.sum())}
s_ref = np.abs(X_an).max(axis=1); al_ref = (s_ref >= np.sort(s_ref)[-k]).astype(int)
iso = IsolationForest(n_estimators=200, contamination=contam, random_state=GRAINE).fit(X_an); s_iso = -iso.score_samples(X_an); al_iso = (iso.predict(X_an) == -1).astype(int)
lof = LocalOutlierFactor(n_neighbors=20, contamination=contam); al_lof = (lof.fit_predict(X_an) == -1).astype(int); s_lof = -lof.negative_outlier_factor_
tab_ano = pd.DataFrame([evalue('Écart maximal (référence)', s_ref, al_ref), evalue('Isolation Forest', s_iso, al_iso), evalue('Local Outlier Factor', s_lof, al_lof)]).set_index('méthode')
RESULTATS['Anomalies (Breast Cancer, 20 malins)'] = tab_ano; display(tab_ano.round(3))
```

## Q7 / Q8 — Renforcement : bandits, Q-learning, SARSA (Gymnasium)

Les trajectoires sont **générées par l'interaction avec un environnement simulé** : il ne s'agit pas d'un jeu de données réel. **Q-learning** et **SARSA** sont comparés sur `FrozenLake-v1` en version glissante (`is_slippery=True`, plus difficile et plus réaliste), sur **5 graines**. Gymnasium ne propose pas d'environnement standard de bandit à bras multiples : la comparaison des **stratégies de bandit** (rotation, ε-greedy, UCB) est donc faite sur une **simulation explicitement signalée**, en NumPy, sur 10 graines. Le **DQN complet est facultatif** et n'est pas réalisé ici ; l'exemple 29 en donne le calcul de cible.

```python
def entrainer(algo, graine, episodes=5000, alpha=.1, gamma=.95, glisse=True):
    env = gym.make('FrozenLake-v1', is_slippery=glisse); n_s, n_a = env.observation_space.n, env.action_space.n
    rng = np.random.default_rng(graine); Q = np.zeros((n_s, n_a)); gains = []
    def choisir_action(s, eps):
        return int(rng.integers(n_a)) if rng.random() < eps else int(rng.choice(np.flatnonzero(Q[s] == Q[s].max())))
    for ep in range(episodes):
        eps = max(.05, 1 - ep / (.6*episodes)); s, _ = env.reset(seed=int(graine*100000 + ep)); a = choisir_action(s, eps); total = 0.
        for _ in range(200):
            s2, r, fin, tronque, _ = env.step(a); total += r; a2 = choisir_action(s2, eps)
            cible = r + (0. if fin else gamma * (Q[s2, a2] if algo == 'sarsa' else Q[s2].max()))
            Q[s, a] += alpha * (cible - Q[s, a]); s, a = s2, a2
            if fin or tronque: break
        gains.append(total)
    env.close(); return Q, np.array(gains)

def taux_reussite(Q, glisse=True, n=500, graine=0):
    env = gym.make('FrozenLake-v1', is_slippery=glisse); succes = 0
    for i in range(n):
        s, _ = env.reset(seed=int(9_000_000 + graine*1000 + i))
        for _ in range(100):
            s, r, fin, tronque, _ = env.step(int(np.argmax(Q[s])))
            if fin or tronque: succes += int(r > 0); break
    env.close(); return succes / n

if gym is None:
    print('gymnasium absent : installez-le (première cellule du notebook) puis relancez.')
else:
    lignes, courbes = [], {}
    for algo in ['qlearning', 'sarsa']:
        reuss, fin_app, cs = [], [], []
        for g in range(5):
            Q, gains = entrainer(algo, g); reuss.append(taux_reussite(Q, graine=g)); fin_app.append(gains[-500:].mean()); cs.append(gains)
        courbes[algo] = np.mean(cs, axis=0)
        lignes.append({'algorithme': 'Q-learning' if algo == 'qlearning' else 'SARSA', 'réussite politique gloutonne (moy.)': np.mean(reuss),
                       'écart-type entre graines': np.std(reuss), 'récompense moyenne des 500 derniers épisodes': np.mean(fin_app)})
    tab_rl = pd.DataFrame(lignes).set_index('algorithme'); RESULTATS['Renforcement (FrozenLake glissant)'] = tab_rl; display(tab_rl.round(3))
    lisse = lambda v, w=200: np.convolve(v, np.ones(w)/w, mode='valid')
    plt.figure(figsize=(6.2, 3.4)); plt.plot(lisse(courbes['qlearning']), label='Q-learning'); plt.plot(lisse(courbes['sarsa']), label='SARSA')
    plt.xlabel('épisode'); plt.ylabel('récompense moyenne (fenêtre de 200)'); plt.title('FrozenLake glissant : courbes d\'apprentissage (moyenne de 5 graines)'); plt.legend(); plt.tight_layout(); plt.show()
```

```python
# Simulation de bandit (NumPy) — PAS des données réelles ; 3 bras de probabilités de gain 0,2 / 0,5 / 0,7 (inconnues de l'agent)
def bandit(strategie, graine, T=2000, vrai=(.2, .5, .7)):
    rng = np.random.default_rng(graine); K = len(vrai); S = np.zeros(K); N = np.zeros(K); total = regret = 0.
    for t in range(T):
        if strategie == 'rotation': a = t % K
        elif strategie == 'epsilon-greedy': a = int(rng.integers(K)) if rng.random() < .1 else int(np.argmax((S + 1) / (N + 1)))
        else: a = t if t < K else int(np.argmax(S / N + np.sqrt(2*np.log(t + 1) / N)))       # UCB1
        r = float(rng.random() < vrai[a]); S[a] += r; N[a] += 1; total += r; regret += max(vrai) - vrai[a]
    return total, regret
lignes = []
for s in ['rotation', 'epsilon-greedy', 'ucb']:
    res = np.array([bandit(s, g) for g in range(10)])
    lignes.append({'stratégie': s, 'récompense cumulée (moy.)': res[:, 0].mean(), 'écart-type': res[:, 0].std(), 'regret cumulé (moy.)': res[:, 1].mean()})
tab_ban = pd.DataFrame(lignes).set_index('stratégie'); RESULTATS['Bandits (simulation NumPy, 10 graines)'] = tab_ban; display(tab_ban.round(2))
```

(2) Pourquoi SARSA et Q-learning peuvent-ils différer (cible utilisant l'action choisie ou la meilleure action) ? (3) Pour les bandits, que montrent le regret et la récompense cumulée sur le compromis exploration / exploitation ?
